# Django Day 1

## Framework

A **framework** is a set of conceptual structures, guidelines, and tools used to build and manage web applications.

- It is a collection of tools and libraries that simplifies the **web development process**.
- It helps developers **speed up development** and maintain **consistency**.

### Features of a Framework

- Reusable code and components
- Consistent structure
- Follows industry standards
- Provides tools for **testing, debugging, and deployment**
- Supports rapid development



## Django Architecture

Django is based on the **MVT (Model–View–Template)** architecture.

There are three main components:

1. **Model**
2. **View**
3. **Template**

### MVT — Model, View, Template

#### 1. Model

- Defines the **structure/schema of the database**.
- Handles data and database-related operations.

#### 2. View

- Contains the **application/business logic**.
- Processes requests and determines the appropriate response.
- Acts as the link between the **Model and Template**.

#### 3. Template

- Contains the **UI/presentation layer** of the application.
- Defines how the data is displayed to the user.
- Usually contains HTML along with Django Template Language (DTL).

### Basic Flow

**User → URL → View → Model → Database**

The View processes the data and sends it to the Template:

**View → Template → Response → User**

# Django Day 2

- **Django** is an open-source, Python-based, high-level web framework that enables rapid web development.
- It follows the **MVT (Model–View–Template)** architecture.

## Features of Django

- Supports rapid development.
- Based on the **MVT architecture**.
- Provides a built-in **ORM (Object-Relational Mapping)**.
- Uses **SQLite3** as the default database.
- Provides a user-friendly **Django Admin** interface.
- Provides excellent documentation.
- Supports **URL routing**.
- Provides **middleware** support.
- Includes a built-in **authentication system**.
- Follows a clean and pragmatic design.
- Promotes **code reusability** and follows the **DRY (Don't Repeat Yourself)** principle.


# System Environment vs Virtual Environment

## 1. System Environment

- It is the **default environment** where the operating system runs and manages installed software and applications.
- Packages installed in the system environment are generally available system-wide.

## 2. Virtual Environment

- It is an **isolated environment** where all project-related dependencies are installed and maintained.
- Modifying dependencies inside a virtual environment does **not affect** the dependencies in the system environment.
- It is recommended to use a **virtual environment (`venv`)** while working on a project.

### Key Difference

**System Environment → Shared/System-wide dependencies**

**Virtual Environment → Project-specific dependencies**

# **Django Day 3**

## **Django Project**

- A **Django project** is a collection of settings and configurations required for a web application.
- It acts as a **container for one or more Django applications**.

### **How to Create a Django Project**

```bash
django-admin startproject ProjectName
```

### **Project Structure**

1. **`__init__.py`**  
   - Indicates that the directory can be treated as a Python package.

2. **`settings.py`**  
   - Contains the configuration settings for the project and its applications.

3. **`urls.py`**  
   - Contains the URL routes/paths of the project.

4. **`wsgi.py` — Web Server Gateway Interface**  
   - Provides an interface between the Django application and a WSGI-compatible web server.
   - Mainly used for **synchronous request handling**.

5. **`asgi.py` — Asynchronous Server Gateway Interface**  
   - Provides an interface between the Django application and an ASGI-compatible server.
   - Supports **asynchronous request handling** and multiple concurrent connections.

---

## **Django Application**

- A **Django application** is a self-contained module that provides a specific functionality or feature within a web application.
- A single Django project can contain **multiple applications**.
- A Django application is generally created inside the **base directory of the project**.

### **How to Create a Django Application**

```bash
python manage.py startapp AppName
```
