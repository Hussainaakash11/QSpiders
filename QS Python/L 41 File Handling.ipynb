## **File Handling**
- A file is a container that stores specific data permanently on a storage device.
- With the help of the file extension, we can identify the type of data stored inside the file:
  - `.py` -> Python file
  - `.txt` -> Text file
  - `.css` -> CSS (Cascading Style Sheets) file
  - `.jpeg` -> Image file
  - `.csv` -> Comma-Separated Values file / spreadsheet
- **File Handling:** It is the process of creating, reading, updating, and writing data inside a file using Python code.
- To read from or write to a file, we first need to open and gain access to the file.
- `open()` -> It is used to open a file and establish a connection/access to it.
- `close()` -> It is used to close the connection with the file and free up system resources.
- `mode` -> It specifies the type of operation we want to perform on the file.
- Syntax:
```python
var = open('file_name_or_path', 'mode')
```

### There are three basic file modes:
1. Write mode (`'w'`)
2. Read mode (`'r'`)
3. Append mode (`'a'`)


### **1. Write Mode (`'w'`):**
- It is denoted by `'w'`.
- It is used to write data inside a file.
- In this mode:
  - If the file does not exist -> It creates a new file.
  - If the file already exists -> It truncates (overwrites) the existing data.
- Methods used to write data:
  - `write()`: Used to write a single string at a time.
  - `writelines()`: Used to write a sequence/iterable of strings (e.g., a list of strings) at a time.
- Syntax:

```python
# Type 1
var = open('file_name_or_path', 'mode')
var.write(data)
var.writeline([data1,data2,---,dataN])
var.close()

# Type 2

with open('file_name_or_path', 'mode') as var:
  var.write(data)
  var.writeline([data1,data2,---,dataN])
  var.close()
```

````python
var = open('vision.txt','w')
var.write('Hello World\n')
var.writelines(['This is a test file.\n', 'This is another line.\n'])
var.close()
````

````python
with open('vision.txt','w') as var:
    var.write('Hello World\n')
    var.writelines(['Talha\n', 'Bella.\n'])
    var.close()
````

### **2. Read Mode (`'r'`):**
- It is denoted by `'r'` (this is also the default mode if no mode is specified).
- It is used to read data from an existing file.
- In this mode:
  - If the file does not exist -> It raises a `FileNotFoundError`.
  - If the file exists -> It opens the file and places the file pointer at the beginning of the file.
- Methods used to read data:
  - `read()`: Used to read the entire content of the file (or a specified number of characters) as a single string.
  - `read(n)`: It read n no. of characters from the file.
  - `readline()`: Used to read a single line from the file at a time.
  - `readlines()`: Used to read all lines from the file and return them as a list of strings.


**1. Using `open()` and `close()`:**
```python
var = open('file_name_or_path', 'r')
data = var.read()        # or var.readline() / var.readlines()
print(data)
var.close()
```

**2. Using `with` statement (Recommended):**

```python
with open('file_name_or_path', 'r') as var:
    data = var.read()    # or var.readline() / var.readlines()
    print(data)
```

````python
var = open('vision.txt','r')
data = var.read()
print(data)
var.close()
````

````python
var = open('vision.txt','r')
data = var.readlines()
print(data)
var.close()
````

### **3. Append Mode (`'a'`):**
- It is denoted by `'a'`.
- It is used to add (append) new data to the end of an existing file without deleting its existing content.
- In this mode:
  - If the file does not exist -> It creates a new file.
  - If the file already exists -> It opens the file and places the file pointer at the very end of the file, preserving existing data.
- Methods used to append data:
  - `write()`: Used to append a single string at the end of the file.
  - `writelines()`: Used to append a sequence/iterable of strings (e.g., a list of strings) at the end of the file.

### **Syntax for Append Mode (`'a'`):**

**1. Using `open()` and `close()`:**
```python
var = open('file_name_or_path', 'a')
var.write("New text to append\n")    # or var.writelines(["Line 1\n", "Line 2\n"])
var.close()

```

**2. Using `with` statement (Recommended):**

```python
with open('file_name_or_path', 'a') as var:
    var.write("New text to append\n")    # or var.writelines(["Line 1\n", "Line 2\n"])
```

````python
var = open('vision.txt','a')
var.write('This is an appended line.\n')
var.close()
````

````python
with open('vision.txt','a') as var:
    var.write('This is another appended line.\n')
    var.close()
````

### **File Handling in `csv` File:**
- It stands for Comma-Separated Values, where data fields are separated using commas.
- Spreadsheet applications like Microsoft Excel support CSV files, making them convenient for exchanging data across different formats and platforms.
- In real-time applications, tabular data is frequently stored and transferred in CSV format.

#### **Writing Data in `csv` Format:**
- In a `csv` file, data is written in a row-by-row format.
- Each list (or iterable) represents a single row.
- `csv.writer()`: Returns a writer object used to convert data into delimited strings and write it to a file object. It is present in the built-in `csv` module.
- `writerow()`: Used to write a single row at a time.
- `writerows()`: Used to write multiple rows (e.g., a list of lists) at a time.

````python
import csv
with open('vision.csv','w',newline='') as var:
    a = csv.writer(var)
    row1 = ['Name','Age','City']
    row2 = ['Talha',25,'Karachi']
    row3 = ['Bella',30,'Lucknow']
    a.writerow(row1)
    a.writerow(row2)
    a.writerow(row3)
    var.close()
````
