### **Iterator**

- An **iterator** is an object that returns **one value at a time** instead of providing all values at once.
- It helps process elements **sequentially** and can be memory-efficient.
- Once an iterator is **exhausted**, it cannot be reused.

### **Iteration**

- **Iteration** is the process of accessing or fetching **one value at a time** from an iterable.

### **Iterable**

- An **iterable** is an object whose elements can be accessed one by one.
- Examples: `str`, `list`, `tuple`, `set`, `dict`, etc.

### **Iterator**

- An **iterator** is an object that performs iteration over an iterable.
- It follows the iterator protocol using `__iter__()` and `__next__()`.

### **Using Iterator on Built-in Objects**

Two built-in functions are commonly used:

1. **`iter()`**
   - Converts an iterable into an iterator.
   - Returns an iterator object.

2. **`next()`**
   - Returns the next value from the iterator.
   - Moves the iterator's position to the next element.
   - Raises `StopIteration` when no elements are left.

```python
# Example
list1 = ['Python',95,2.4,"java",False]
iter_obj=iter(list1)

print(next(iter_obj))
print(next(iter_obj))
print(next(iter_obj))
print(next(iter_obj))
print(next(iter_obj))
# print(next(iter_obj)) ---> Raises: StopIteration
```

### **Using Iterator on a User-Defined Object**

- To make a user-defined object behave as an **iterator**, we commonly implement two magic methods:
  1. `__iter__()`
  2. `__next__()`

### **1. `__iter__()`**

- It returns the **iterator object**.
- It is called when `iter()` is applied to the object or when iteration begins.
- Usually, an iterator returns `self` from `__iter__()`.

### **2. `__next__()`**

- It returns the **next value** from the iterator.
- It is called automatically during each iteration.
- It should raise `StopIteration` when there are no more values.

```python
class Range:
    def __init__(self,val):
        self.max_val=val
        self.count=0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < self.max_val:
            self.count += 1
            return self.count

        else:
            raise StopIteration

obj=Range(12)

for i in obj:
    print(i)
```

### **Generator**

- A **generator** is a simpler way to create an **iterator**.
- A generator is a function that uses the **`yield`** keyword to produce values one at a time.
- If a function contains `yield`, it becomes a **generator function**.
- Calling a generator function returns a **generator object**, which is an iterator.
- Values can be accessed using a **`for` loop** or the `next()` function.
- Generators are **memory-efficient** because they generate values one at a time instead of storing all values in memory.

### **`yield`**

- `yield` is a keyword used to **return a value temporarily** from a generator function.
- Unlike `return`, `yield` **pauses the function and preserves its state**.
- The function resumes execution from where it was paused when the next value is requested.

```python
def demo():
    yield 10
    yield 20
    yield 30

var = demo()
# for i in var:
#     print(i)
print(type(var))
# print(next(var),next(var),next(var))
```

```python
def palindrome(N):
    for i in range(1,N+1):
        if str(i)==str(i)[::-1]:
            yield i

N=int(input("Enter N: "))

var = palindrome(N)
for i in var:
    print(i)
```
