## **Decorators**

- A **decorator** is used to add extra functionality to a function **without modifying its original code**.
- It is applied to a function using the `@` symbol.
- A decorator is essentially a function that **takes another function as an argument and returns a new function**.

### **Types of Decorators**

There are two main types of decorators:

#### **1. Built-in Decorators**

Python provides several built-in decorators, such as:

- `@classmethod`
- `@staticmethod`
- `@property`
- `@abstractmethod`

#### **2. User-Defined Decorators**

- These are decorators **created by the programmer according to specific requirements**.
- A user-defined decorator generally **accepts a function as an argument and returns a wrapper function**.
- It is commonly implemented using a **nested function**.

### **Basic Syntax**

```python
# How to define a decorator?

def decorator_name(func):

    def wrapper(*args,**kwargs):
        # pre task
        func(*args,**kwargs)
        # post task

    return wrapper

# How to use a decorator?

@decorator_name
def func_name():
    # Statement Block
    print("Hello")

func_name()

# How decorator is called internally?

func_name=decorator_name(func_name)
```

Here, `wrapper()` adds extra functionality to `display()` without modifying the original `display()` function.

````python
def instagram(func):
    def wrapper(*args, **kwargs):
        print("Go To https://www.instagram.com/")
        print("Log in to your Instagram account.")
        func(*args, **kwargs)
        print("Log out of your Instagram account.")
    return wrapper

def aman():
    print("Posted a story on Instagram")

@instagram
def shikhar():
    print("Follows Girls")

def nirmal():
    print("Watches Reels")

shikhar()
````
