## **`for` Loop**

* A `for` loop is used to **iterate over the elements of an iterable**.
* It is useful when the number of iterations is known or when we need to process each element of a collection.
* **Initialization and updation are handled automatically** by the `for` loop.
* It works with iterable data types such as `str`, `list`, `tuple`, `set`, and `dict`.
* The loop variable takes the value of each element **one by one**.
* For a collection, the number of iterations generally depends on the **number of elements**.

### **Syntax**

```python
for var in iterable:
    statement
```

### **Example**

```python
name = "Python"

for char in name:
    print(char)
```
