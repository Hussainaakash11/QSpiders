## **Abstraction**

- **Abstraction** is the process of **hiding implementation details** and exposing only the required functionality to the user.
- It helps define the **interface, design, or blueprint** for a product/class.
- In Python, abstraction can be implemented using three main components:

1. **Abstract Class** → Defines the interface/blueprint.
2. **Abstract Method** → Defines the required features or functionalities.
3. **Concrete Class** → Provides the actual implementation of those features.

> **Note:** Abstraction ensures that the required features defined by the abstract class (Interface) are implemented by the concrete class (Child).

### **1. Abstract Class**

- An **abstract class** is a class used to define an **interface or blueprint** for its child classes.
- It can contain **abstract methods** as well as concrete methods.
- To create an abstract class, inherit from `ABC`, which is provided by Python's **`abc` module**.
- An abstract class **cannot be instantiated directly**.

### **2. Abstract Method**

- An **abstract method** is a method that is **declared but not implemented** in the abstract class.
- It is used to define the **required features or functionalities** that child classes must implement.
- The `@abstractmethod` decorator is used to define an abstract method.
- `abstractmethod` is provided by Python's **`abc` module**.

### **Syntax**

```python
from abc import ABC, abstractmethod

class ClassName(ABC):

    @abstractmethod
    def method_name(self):
        pass
```

### **3. Concrete Class**

- A **concrete class** represents the actual **product or implementation**.
- It is a **child class of the abstract class**.
- It must **implement/override all abstract methods** defined in the abstract class.
- Objects **can be created** from a concrete class.

### **Example**

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car starts")

obj = Car()
obj.start()
```

Here, `Car` is a concrete class because it provides the implementation of the `start()` abstract method.

````python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def light_check(self):
        pass

    @abstractmethod
    def start(self):
            pass

    @abstractmethod
    def engine_check(self):
        pass

    @abstractmethod
    def lock_check(self):
            pass

    @abstractmethod
    def wiper_check(self):
            pass

class Thar(Vehicle):

    
    def light_check(self):
        print("Light is OK")

    def lock_check(self):
        print("Lock is OK")

    def start(self):
        print("Engine Started")

    def engine_check(self):
        print("Engine check successful")

    def wiper_check(self):
        print("Wiper is OK")


obj=Thar()
obj.start()
````
