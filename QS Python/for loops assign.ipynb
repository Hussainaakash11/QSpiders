### 1. Extract list of even numbers and list of odd numbers from a given list.

````python
l1=[2,3,4,5,6,7,55,44,33,22,66]
l2=[]
l3=[]
for num in l1:
    if num%2==0:
        l2.append(num)
    else:
        l3.append(num)
print("Org List",l1)
print("Even List",l2)
print("Odd List",l3)
````

### 3. Extract all the palindrome numbers from a list.

````python
l=[11,22,33,45,567,664,343,456,6534,345]
final=[]
for i in l:
    if str(i)==str(i)[::-1]:
        final.append(i)
print(final)
````

### 4. Get all the uppercase, lowercase, special characters and numbers in different lists.
Given:
```python
s = 'ADmin@#123.com'

l1 = ['A', 'D']
l2 = ['m', 'i', 'n']
l3 = ['@', '#', '.']
l4 = ['1', '2', '3']
```

````python
s='ADmin@#123.com'
l1=[]
l2=[]
l3=[]
l4=[]
for i in s:
    if 'A'<=i and 'Z'>=i:
        l1.append(i)
    elif 'a'<=i and 'z'>=i:
        l2.append(i)
    elif '0'<=i and '9'>=i:
        l3.append(i)
    else:
        l4.append(i)
print("Original data: ",s)
print(l1)
print(l2)
print(l3)
print(l4)
````

### 5. Add all the values present in even index positions and find the product of values present in odd index positions.

````python
list=[1,2,3,5,67,8,9,3]
add=0
prod=1
for i in range(len(list)):
    if i%2==0:
        add+=list[i]
    else:
        prod=prod*list[i]
print("sum",add)
print("Product",prod)
````
