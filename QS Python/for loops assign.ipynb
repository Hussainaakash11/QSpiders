### 1. Extract list of even numbers and list of odd numbers from a given list.

````python
l1=[2,3,4,5,6,7,55,44,33,22,66]
l2=[]
l3=[]
for num in l1:
    if num%2==0:
        l2.append(num)
    else:
        l3.append(num)
print("Org List",l1)
print("Even List",l2)
print("Odd List",l3)
````

### 3. Extract all the palindrome numbers from a list.

````python
l=[11,22,33,45,567,664,343,456,6534,345]
final=[]
for i in l:
    if str(i)==str(i)[::-1]:
        final.append(i)
print(final)
````

### 4. Get all the uppercase, lowercase, special characters and numbers in different lists.
Given:
```python
s = 'ADmin@#123.com'

l1 = ['A', 'D']
l2 = ['m', 'i', 'n']
l3 = ['@', '#', '.']
l4 = ['1', '2', '3']
```

````python
s='ADmin@#123.com'
l1=[]
l2=[]
l3=[]
l4=[]
for i in s:
    if 'A'<=i and 'Z'>=i:
        l1.append(i)
    elif 'a'<=i and 'z'>=i:
        l2.append(i)
    elif '0'<=i and '9'>=i:
        l3.append(i)
    else:
        l4.append(i)
print("Original data: ",s)
print(l1)
print(l2)
print(l3)
print(l4)
````

### 5. Add all the values present in even index positions and find the product of values present in odd index positions.

````python
list=[1,2,3,5,67,8,9,3]
add=0
prod=1
for i in range(len(list)):
    if i%2==0:
        add+=list[i]
    else:
        prod=prod*list[i]
print("sum",add)
print("Product",prod)
````

### 6. Count the even digits and odd digits. 

````python
num = int(input("Enter a number: "))

even = 0
odd = 0

while num > 0:
    digit = num % 10

    if digit % 2 == 0:
        even += 1
    else:
        odd += 1

    num //= 10

print("Even digits:", even)
print("Odd digits:", odd)
````

### 7. Count the digits greater than 5.

````python
num = int(input("Enter a number: "))

count = 0

while num > 0:
    digit = num % 10

    if digit > 5:
        count += 1

    num //= 10

print("Digits greater than 5:", count)
````

### 8. Count the digits that are multiples of 3.

### 9. Count the number of vowels in a string.

````python
s = input("Enter a string: ")

count = 0
i = 0

while i < len(s):
    if s[i] in "aeiouAEIOU":
        count += 1

    i += 1

print("Total vowels:", count)
````

### 10. Count the number of consonants in a string.

````python
s = input("Enter a string: ")

count = 0
i = 0

while i < len(s):
    if s[i].isalpha() and s[i] not in "aeiouAEIOU":
        count += 1

    i += 1

print("Total consonants:", count)
````

### 11. Count the number of special symbols in a string.

````python
s = input("Enter a string: ")

count = 0
i = 0

while i < len(s):
    if not s[i].isalnum():
        count += 1

    i += 1

print("Total special symbols:", count)
````

### 12. Count palindrome numbers in a range.

````python
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

count = 0

while start <= end:

    num = start
    rev = 0

    while num > 0:
        digit = num % 10
        rev = rev * 10 + digit
        num //= 10

    if start == rev:
        count += 1

    start += 1

print("Total palindrome numbers:", count)
````

### 13. Extract the special symbols from a given string

````python
s = input("Enter the string")

sp = ''

for i in s:

    if i != " " and not (('A' <= i <= 'Z') or
                         ('a' <= i <= 'z') or
                         ('0' <= i <= '9')):
        sp += i

print(sp)
````

### 14. Toggle a string

````python
s=input("Enter a string: ")
out=''
for i in s:
    if 'A' <= i <= 'Z':
        out+=chr(ord(i)+32)
    elif 'a' <= i <= 'z':
        out+=chr(ord(i)-32)
    else:
        out += i
print(out)
````
