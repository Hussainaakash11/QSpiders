# **Looping Statements**

* Looping statements are used to **execute a set of instructions repeatedly**.
* Python provides two main types of loops:

  1. `while` loop
  2. `for` loop

## **`while` Loop**

* A `while` loop repeatedly executes a block of statements **as long as the given condition is `True`**.
* The loop terminates when the condition becomes `False`.
* **Initialization** and **updation** are generally required to control the loop.
* The **updation must be performed inside the loop body** to avoid an infinite loop.

### **Syntax**

```python
i = 1              # Initialization

while i < 10:      # Condition
    # True Statement Block
    print(i)
    i = i + 1      # Updation

# Statements after the loop
```

> **Note:** Fetch the values from a collection using while loop uses this syntax given below:
```python
while i<len(collection):

or 

while i <= len(collection)-1:

```

## **Nested `while` Loop**

* A **nested `while` loop** is a `while` loop placed inside the **statement block of another `while` loop**.
* The outer loop controls the execution of the inner loop.
* For every iteration of the outer loop, the inner loop executes completely.

### **Syntax**

```python
i = 1

while i <= 3:
    j = 1

    while j <= 3:
        # Inner loop statement block
        print(i, j)
        j = j + 1

    i = i + 1
```

````python
# Factorial of N natural numbers

n = int(input("Enter n: "))

i = 1

while i <= n:
    fact = 1
    j = 1

    while j <= i:
        fact = fact * j
        j = j + 1

    print("Factorial of", i, "=", fact)
    i = i + 1
````
