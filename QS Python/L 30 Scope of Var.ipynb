# **Scope of Variables**

- **Scope of a variable** defines the region of a program where the variable can be **accessed and used**.
- Python follows the **LEGB rule** to determine the scope of a variable:

**L → E → G → B**
- **L** → Local
- **E** → Enclosed
- **G** → Global
- **B** → Built-in

## **1. Local Scope**

- A variable declared **inside a function** has local scope.
- It can be accessed **only within that function**.

```python
def display():
    x = 10
    print(x)

display()
```

Here, `x` is a **local variable**.


````python
def demo():
    x=200
    print(x)
demo()
````

## **2. Enclosed / Non-Local Scope**

- It occurs when **functions are nested**.
- A variable defined in the **outer function** and accessed by an inner function has an enclosed scope.
- The `nonlocal` keyword is used to **modify an enclosed variable from the inner function**.


Here, `x` is an **enclosed variable** for the `inner()` function.

````python
def outer():
    x = 10

    def inner():
        nonlocal x
        x = 20

    inner()
    print(x)

outer()
````

## **3. Global Scope**

- A variable declared **outside all functions** has global scope.
- It can be accessed from the **global space and inside functions**.
- The `global` keyword is used to **modify a global variable from inside a function**.

````python
x = 10

def display():
    global x
    x = 20

display()
print(x)
````

## **4. Built-in Scope**

- Built-in scope contains Python's **predefined names** that are available throughout a program.
- Examples: `print()`, `input()`, `len()`, `True`, `False`, `int`, etc.


Here, `print()` and `len()` are **built-in names**.

````python
print(len("Python"))
````

### **LEGB Rule**

Python searches for a variable in the following order:

**Local → Enclosed → Global → Built-in**

```text
L → E → G → B
```

Python stops searching as soon as it finds the required name.
