## **Comprehension**

- **Comprehension** is a concise way of creating a new collection using a **single-line expression**.
- It can include **iteration and conditions** to generate elements efficiently.
- Types of comprehensions:
  1. **List Comprehension**
  2. **Set Comprehension**
  3. **Dictionary Comprehension**
  4. **Generator Expression**

### **1. List Comprehension**

- **List comprehension** is a concise way of creating a new list using a **single-line expression**.
- It can include **loops and conditions** to generate list elements.

### **Syntax**

```python
var = [i for i in iterable] 
# Simple Loop

var = [i for i in iterable if condition]
# One Statement Block

var = [TSB if condition else FSB for i in iterable]
# Two Statement Blocks

var = [(i, j) for i in iterable1 for j in iterable2]
# Nested Loop
```

Here:
- `TSB` → True Statement Block
- `FSB` → False Statement Block

````python
# List of N natural Numbers
n=int(input("Enter N"))
s = [i for i in range(1,n+1)]
print(s)
````

````python
# List of N natural Palindrom Numbers
n=int(input("Enter N"))
o= [i for i in range(1,n+1) if str(i)==str(i)[::-1]]
print(o)
````

````python
# print sq if even else cube
t1=(1,2,3,4,5,6)
l1=[i**2 if i%2==0 else i**3 for i in t1]
print(l1)
````

````python
# Flattern the list.
l1=[[1,2],[3,4],[5,6]]
res = [j for i in l1 for j in i ]
print(res)
````

### **2. Set Comprehension**

- **Set comprehension** is a concise way of creating a new set using a **single-line expression**.
- It can include **loops and conditions** to generate set elements.

### **Syntax**

```python
var = {i for i in iterable} 
# Simple Loop

var = {i for i in iterable if condition}
# One Statement Block

var = {TSB if condition else FSB for i in iterable}
# Two Statement Blocks

var = {(i, j) for i in iterable1 for j in iterable2}
# Nested Loop
```

Here:
- `TSB` → True Statement Block
- `FSB` → False Statement Block

````python
# List of N natural Numbers
n=int(input("Enter N"))
s = {i for i in range(1,n+1)}
print(s)
````

````python
# print sq if even else cube
t1=(1,2,3,4,5,4,5,6,7,8,6)
s1={i**2 if i%2==0 else i**3 for i in t1}
print(s1)
````

````python
# List of N natural Palindrom Numbers
n=int(input("Enter N"))
o= {i for i in range(1,n+1) if str(i)==str(i)[::-1]}
print(o)
````

### **3. Dictionary Comprehension**

- **Dict comprehension** is a concise way of creating a new Dict using a **single-line expression**.
- It can include **loops and conditions** to generate dict elements.

### **Syntax**

```python
var = {key:val for i in iterable} 
# Simple Loop

var = {key:val for i in iterable if condition}
# One Statement Block

var = {key:TSB if condition else FSB for key:val in iterable}
# Two Statement Blocks

var = {key:val for i in iterable1 for j in iterable2}
# Nested Loop
```

Here:
- `TSB` → True Statement Block
- `FSB` → False Statement Block

````python
# Dict of N natural Numbers
s = {key:key*10 for key in range(1,int(input("Enter N"))+1)}
print(s)
````

````python
# l1 = ['hii',10,'hello',2.5,9-4j,'bye']
# o/p = {'hii':3, 'hello':5, 'bye':3}

l1 = ['hii',10,'hello',2.5,9-4j,'bye']
d1 = {i:len(i) for i in l1 if type(i)==str}
print(d1)
````

````python
l1 = ['hii',10,'hello',2.5,9-4j,'bye']
d1 = {i:i[::-1] for i in l1 if type(i)==str}
print(d1)
````

````python
n=5
d1={i:i**3 if i%2 !=0 else i**2 for i in range(1,n+1)}
print(d1)
````

````python
# o/p {'hii@':'@' , 'bye$': '$' , 'he#llo':'#'}
l1 = ['hii@', 'bye$','he#llo']
# d1={i:val for i in l1 for val in i if val in '!@#$%^&*()_+='}
d1={i:val for i in l1 for val in i if not(val.isalnum())}
print(d1)
````

### **Generator Expression**

- **Generator Expression** is a concise way of creating a Generator using a **single-line expression**.
- It can include **loops and conditions** to generate Generator elements.

### **Syntax**

```python
var = (i for i in iterable) 
# Simple Loop

var = (i for i in iterable if condition)
# One Statement Block

var = (key:TSB if condition else FSB for i in iterable)
# Two Statement Blocks

var = (i:j for i in iterable1 for j in iterable2)
# Nested Loop
```

Here:
- `TSB` → True Statement Block
- `FSB` → False Statement Block
