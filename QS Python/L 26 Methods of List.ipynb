# **Built-in List Methods**

Python provides **11 built-in methods** for lists:

| No. | Method | Purpose | Syntax |
|---:|---|---|---|
| 1 | `append()` | Adds an element at the end of the list. | `var.append(value)` |
| 2 | `extend()` | Adds multiple elements to the end of the list. | `var.extend(iterable)` |
| 3 | `insert()` | Inserts an element at a specified index. | `var.insert(index, value)` |
| 4 | `remove()` | Removes the first occurrence of a specified value. | `var.remove(value)` |
| 5 | `pop()` | Removes and returns an element from the specified index. | `var.pop(index)` |
| 6 | `clear()` | Removes all elements from the list. | `var.clear()` |
| 7 | `index()` | Returns the index of the first occurrence of a value. | `var.index(value)` |
| 8 | `count()` | Returns the number of occurrences of a value. | `var.count(value)` |
| 9 | `sort()` | Sorts the list in ascending order by default. | `var.sort()` |
| 10 | `reverse()` | Reverses the order of elements in the list. | `var.reverse()` |
| 11 | `copy()` | Returns a shallow copy of the list. | `var.copy()` |
