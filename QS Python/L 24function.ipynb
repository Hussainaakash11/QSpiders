## **Functions**

* A **function** is a reusable block of code designed to perform a **specific task**.
* Functions help divide a program into **small, manageable units**.
* A function can accept **inputs (parameters)** and may return an **output (value)**.
* Functions are defined using the `def` keyword and executed by **calling the function**.

### **Applications**

1. **Reusability:** Write code once and use it multiple times.
2. **Readability:** Makes the program easier to understand.
3. **Structure:** Provides a clear structure to the program.
4. **Maintainability:** Makes code easier to modify, debug, and maintain.
5. **Modularity:** Divides a large program into smaller, independent modules.

## **Types of Functions**

There are two main types of functions in Python:

### **1. Built-in Functions**

* **Built-in functions** are predefined functions provided by Python.
* They are already implemented and can be directly used in a program.
* We can **call and use** them, but we generally do not modify their built-in implementation.

### **Types of Built-in Functions**

1. **Utility Functions:**
   `id()`, `len()`, `type()`, `print()`, `input()`, etc.

2. **String Functions/Methods:**
   Functions or methods used to perform operations on strings.

3. **List Functions/Methods:**
   Functions or methods used to perform operations on lists.

4. **Tuple Functions/Methods:**
   Functions or methods used to perform operations on tuples.

5. **Set Functions/Methods:**
   Functions or methods used to perform operations on sets.

6. **Dictionary Functions/Methods:**
   Functions or methods used to perform operations on dictionaries.

> **Note:** In Python, operations such as `str.upper()`, `list.append()`, and `dict.keys()` are technically called **methods**, because they belong to specific objects/types. Functions like `len()` and `type()` are built-in functions.


## **Built-in Methods of String:**

String methods are **built-in methods used to perform various operations on string values**.

### **1. `upper()`**

Converts all alphabetic characters of a string to uppercase.

**Syntax:** `var.upper()`

### **2. `lower()`**

Converts all alphabetic characters of a string to lowercase.

**Syntax:** `var.lower()`

### **3. `swapcase()`**

Converts uppercase characters to lowercase and lowercase characters to uppercase.

**Syntax:** `var.swapcase()`

### **4. `capitalize()`**

Converts the first character to uppercase and the remaining alphabetic characters to lowercase.

**Syntax:** `var.capitalize()`

### **5. `title()`**

Converts the first character of each word to uppercase and the remaining alphabetic characters to lowercase.

**Syntax:** `var.title()`

### **6. `isupper()`**

Checks whether all alphabetic characters in the string are uppercase.

**Syntax:** `var.isupper()`

### **7. `islower()`**

Checks whether all alphabetic characters in the string are lowercase.

**Syntax:** `var.islower()`

### **8. `istitle()`**

Checks whether the string follows title-case formatting.

**Syntax:** `var.istitle()`

### **9. `isdigit()`**

Checks whether all characters in the string are digits.

**Syntax:** `var.isdigit()`

### **10. `isalpha()`**

Checks whether all characters in the string are alphabetic.

**Syntax:** `var.isalpha()`

### **11. `isalnum()`**

Checks whether all characters in the string are alphanumeric (letters or digits).

**Syntax:** `var.isalnum()`

### **12. `isspace()`**

Checks whether all characters in the string are whitespace characters.

**Syntax:** `var.isspace()`

### **13. `startswith()`**

Checks whether the string starts with the specified substring.

**Syntax:** `var.startswith(value)`

### **14. `endswith()`**

Checks whether the string ends with the specified substring.

**Syntax:** `var.endswith(value)`


```python
s="KashMir is beautiful"
print("Title:", s.title())
print("Upper:", s.upper())
print("Lower:", s.lower())
print("Capitalize:", s.capitalize())
print("Swapcase:", s.swapcase())
print("Count of 'i':", s.count('i'))
print("Is upper:", s.isupper())
print("Is lower:", s.islower())
print("Is title:", s.istitle())
print("Is alpha:", s.isalpha())
print("Is alphanumeric:", s.isalnum())
print("Is digit:", s.isdigit())
print("Is space:", s.isspace())
print("Starts with 'Kash':", s.startswith('Kash'))
print("Ends with 'beautiful':", s.endswith('beautiful'))
```

### 1. Check whether the given character is a special character or not

```python
ch = input("Enter a character: ")

if ch.isalnum():
    print("Not a special character")
else:
    print("Special character")
```

### 2. Extract uppercase, lowercase, numbers and special characters into different lists

```python
s = 'ADmin@123'

upper = []
lower = []
numbers = []
special = []

for ch in s:
    if ch.isupper():
        upper.append(ch)
    elif ch.islower():
        lower.append(ch)
    elif ch.isdigit():
        numbers.append(ch)
    else:
        special.append(ch)

print("Uppercase:", upper)
print("Lowercase:", lower)
print("Numbers:", numbers)
print("Special characters:", special)
```

### 3. Extract palindrome words from a list and keep them in uppercase in another list

```python
l1 = ['hello', 'level', 'madam', 'python']

out = []

for word in l1:
    if word == word[::-1]:
        out.append(word.upper())

print(out)
```
