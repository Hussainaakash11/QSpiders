## **Arguments and Their Types**

### **1. Parameters**

- Parameters are the **variables defined in the function definition**.
- They are used to **receive the values/arguments** passed during the function call.
- Parameters are also known as **formal parameters**.

```python
def add(a, b):
    print(a + b)
```

Here, `a` and `b` are **parameters**.

---

### **2. Arguments**

- Arguments are the **actual values passed to a function during the function call**.
- Arguments are also known as **actual arguments**.
- Python provides **5 types of arguments**:

1. **Positional Arguments**
2. **Keyword Arguments**
3. **Default Arguments**
4. **Variable-Length Arguments**
5. **Variable-Length Keyword Arguments**

## **1. Positional Arguments**

- Arguments are passed **according to the position of the parameters**.
- The **order and position of arguments matter**.
- The **number of arguments must match the number of parameters**.
- The first argument is assigned to the first parameter, the second to the second, and so on.

### **Syntax**

```python
def func_name(var1, var2, ..., varn):
    # Statement Block

func_name(val1, val2, ..., valn)
```

````python
def demo(a,b,c):
    print(a)
    print(b)
    print(c)

demo(10,20,30)
````

## **2. Keyword Arguments**

- Arguments are passed by **explicitly specifying the parameter name (Keys)**.
- The **order of arguments does not matter**.
- The **parameter names must match** the names defined in the function.
- The number of arguments must generally match the number of required parameters.

### **Syntax**

```python
def func_name(var1, var2, ..., varn):
    # Statement Block

func_name(var1=val1, var2=val2, ..., varn=valn)
```

> Here, the order does not matter because the values are passed using **parameter names**.

````python
def student(name, age):
    print(name, age)

student(age=22, name="Aakash")
````

````python
# Using both positional and keyword arguments
def student_info(name, age, city):
    print(name, age, city)

student_info("Aakash", city="Kashmir", age=21)

# Positional arguments must appear before keyword arguments in a function call. 
# If you try to pass a positional argument after a keyword argument, it will result in a syntax error.
````

## **3. Default Arguments**

- A **default value** is assigned to a parameter in the function definition.
- If an argument is **not provided** during the function call, the default value is used.
- If an argument is provided, it **overrides the default value**.
- Default parameters must be placed **after non-default parameters**.
- Number of parameters **<=** Number of Arguments.

### **Syntax**

```python
def func_name(var1, var2=default_value):
    # Statement Block

func_name(val1)
```

````python
def student(name, age=22):
    print(name, age)

student("Aakash")
````

````python
def person(name,age,email="NA"):
    print(name, age, email)

print(person("Aakash", 22 ))
print(person("Aakash", 22, "aakash@example.com"))
````

## **4. Variable-Length Arguments**

- Variable-length arguments are used when the **number of arguments is not fixed**.
- They allow a function to accept **any number of positional arguments**.
- The `*` symbol is used before the parameter name.
- Inside the function, the arguments are received as a **tuple**.

### **Syntax**

```python
def func_name(*args):
    # Statement Block

func_name(val1, val2, ..., valn)
```

Here, `*numbers` can accept any number of positional arguments, and the values are stored as a **tuple**.

````python
def add(*numbers):
    num=0
    for i in numbers:
        num=num+i
    print(num)

add(10, 20, 30, 40)
````

## **5. Variable-Length Keyword Arguments**

- Variable-length keyword arguments are used when the **number of keyword arguments is not fixed**.
- They allow a function to accept **any number of keyword arguments**.
- The `**` symbol is used before the parameter name.
- Inside the function, the arguments are received as a **dictionary**.
- Each argument is stored as a **key-value pair**.

### **Syntax**

```python
def func_name(**kwargs):
    # Statement Block

func_name(key1=val1, key2=val2, ..., keyN=valN)
```

Here, `**details` can accept any number of keyword arguments, and the values are stored as a **dictionary**.

````python
def student(**details):
    print(details)

student(name="Aakash", age=22, branch="IT")
````
