## **Encapsulation**

- **Encapsulation** is the process of **wrapping data and related functionalities into a single unit**, such as a class.
- It helps **control access to data** and provides better data security using **access modifiers**.
- In Python, access is mainly controlled through naming conventions.

### **Types of Access Modifiers**

1. **Public Members**
2. **Private Members**

### **1. Public Members**

- **Public members** are class members that can be accessed **inside the class, outside the class, and in child classes**.
- In Python, class members are **public by default** unless a naming convention is used to indicate restricted access.

### **Example**

```python
class Student:
    name = "Aakash"

obj = Student()

print(obj.name)  # Accessible outside the class
```

### **2. Private Members**

- **Private members** are class members intended to be accessed **only within the class**.
- They are not directly accessible from outside the class or through a child class.
- A **double underscore (`__`)** before the member name is used to define it as private.
- Private members help **restrict direct access to data**.

### **Example**

```python id="r6q2kw"
class Student:
    def __init__(self):
        self.__name = "Aakash"

    def show(self):
        print(self.__name)

obj = Student()
obj.show()
```

Here, `__name` is a private member and cannot be directly accessed as `obj.__name`.

````python
class Bank:
    def __init__(self,name,bal):
        self.name=name
        self.__bal=bal

    def get_bal(self):
        password=input("Enter Password")
        if password=="Chick":
            return self.__bal
        else:
            print("Access Deined")

p1 = Bank("virat",15000)
print(p1.name)
print(p1.get_bal())
````

### **Property Decorator**

- `@property` is a **built-in decorator** that allows a method to be accessed like a **property**, without using parentheses `()`.
- It is commonly used to **access private members** through a getter method.
- It provides controlled access to private data.
- It has two commonly used attributes:
  1. **Setter** → Used to update the value.
  2. **Deleter** → Used to delete the value.

````python
class Student:
    def __init__(self, name):
        self.name = name

    @property
    def age(self):
        return 21

obj = Student("vivek")

print(obj.name)
````

````python
class Bank:
    def __init__(self,name,bal):
        self.name=name
        self.__bal=bal

    @property
    def bal(self):
        password=input("Enter Password")
        if password=="Chick":
            return self.__bal
        else:
            print("Access Deined")

p1 = Bank("virat",15000)
print(p1.name)
print(p1.bal)
````

````python
class Bank:
    def __init__(self,name,bal):
        self.name=name
        self.__bal=bal

    @property
    def bal(self):
        password=input("Enter Password")
        if password=="Chick":
            return self.__bal
        else:
            print("Access Deined")

    @bal.setter
    def bal(self,new):
        self.__bal = new

    @bal.deleter
    def bal(self):
        del self.__bal

p1 = Bank("virat",15000)
print(p1.name)
print(p1.bal)

p1.bal=2000000

print("Access Modifier =====>")

print(p1.bal)

del p1.bal
print("Balance is successfully Deleted")

print(p1.bal)
````
