## **Copy Operations**

* Copy operation is the process of creating a **copy/reference of existing data**.
* It is commonly used to preserve the original data while working with a copy.
* There are **three types** of copy operations:

  1. General Copy
  2. Shallow Copy
  3. Deep Copy

### **1. General Copy**

* Both variables **refer to the same memory location/object**.
* Changes made through one variable **affect the other variable**.
* It does **not create a separate copy** or require additional memory for the object.

**Syntax:**

```python
new_var = old_var
```

**Example:**

```python
l1 = [10, 20, 30]

l2 = l1
l2[1] = 6

print(l1)  # [10, 6, 30]
print(l2)  # [10, 6, 30]
```

> **Note:** `l2 = l1` creates a **reference**, not an independent copy.


````python
l1 = [10, 20, 30]

l2 = l1
l2[1] = 6

print(l1)  # [10, 6, 30]
print(l2)  # [10, 6, 30]
````

### **2. Shallow Copy**

* In a **shallow copy**, the original and copied variables refer to **different objects/memory locations**.
* Changes to the **new elements** of the copied object do not affect the original.
* For nested objects, the inner objects may still be **shared**.

### **Ways to Perform Shallow Copy**

#### **1. Using Slicing**

**Syntax:**

```python
new_var = original[::]
```

**Example:**

```python
l1 = [10, 20, 30]

l2 = l1[::]
l2[1] = 6

print(l1)  # [10, 20, 30]
print(l2)  # [10, 6, 30]
```

#### **2. Using `copy()`**

**Syntax:**

```python
import copy

new_var = copy.copy(original_var)
```

**Example:**

```python
import copy

l1 = [10, 20, 30]
l2 = copy.copy(l1)

l2[1] = 6

print(l1)  # [10, 20, 30]
print(l2)  # [10, 6, 30]
```
### **Disadvantage of Shallow Copy**

* In a shallow copy, the **outer object** has a different memory address, but **nested/inner objects are still shared** between the original and copied objects.
* Therefore, modifying a nested object can **affect the original data**.

````python
l1 = [10, 20, 30]

l2 = l1[::]
l2[1] = 6

print(l1)  # [10, 20, 30]
print(l2)  # [10, 6, 30]
````

````python
import copy

l1 = [10, 20, 30]
l2 = copy.copy(l1)

l2[1] = 6

print(l1)  # [10, 20, 30]
print(l2)  # [10, 6, 30]
````

````python
# Example of shallow copy Disadvantage
l1 = [[10, 20], 30, 40]
l2 = copy.copy(l1)
print("Before modification:")
print(l1)
print(l2) 

l2[0][1] = 6
print("After modification:")
print(l1)  
print(l2)  
````
