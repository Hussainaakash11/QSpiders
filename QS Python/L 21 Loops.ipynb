# **Looping Statements**

* Looping statements are used to **execute a set of instructions repeatedly**.
* Python provides two main types of loops:

  1. `while` loop
  2. `for` loop

## **`while` Loop**

* A `while` loop repeatedly executes a block of statements **as long as the given condition is `True`**.
* The loop terminates when the condition becomes `False`.
* **Initialization** and **updation** are generally required to control the loop.
* The **updation must be performed inside the loop body** to avoid an infinite loop.

### **Syntax**

```python
i = 1              # Initialization

while i < 10:      # Condition
    # True Statement Block
    print(i)
    i = i + 1      # Updation

# Statements after the loop
```
