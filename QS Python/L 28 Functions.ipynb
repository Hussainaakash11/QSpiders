# **User Defined Functions**

- A **user-defined function** is a function created by the programmer according to specific requirements.
- It contains the **business or application logic** required to perform a particular task.
- The `def` keyword is used to define a function.
- A function is **not executed until it is called**.
- The name of the function should be in **lower case**
- A function can accept **parameters** and may return a value using the `return` statement.

### **Syntax**

```python
def func_name(var1, var2, ..., varn):
    # Business Logic
    statement(s)

func_name(val1, val2, ..., valn)
```

````python
def add(a, b):
    result = a + b
    print(result)

add(10, 20)
````

### **Parameters**
- Parameters are the **variables defined in the function definition**.
- They receive values when the function is called.

```python
def add(a, b):   # a and b are parameters
    print(a + b)
```

### **Arguments**
- Arguments are the **actual values passed to a function when it is called**.

```python
add(10, 20)      # 10 and 20 are arguments
```

**In short:**  
**Parameter → Variable in function definition**  
**Argument → Actual value passed during function call**

---

### **`return` Keyword**
- `return` is used to **send a value back from a function to the calling statement**.
- It also **terminates the function execution**.
- The returned value can be stored in a variable or used in an expression.


**Note:** If a function does not explicitly use `return`, it returns `None` by default.

````python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
````

## **Ways of Defining a Function**

Functions can be categorized based on whether they **accept parameters** and **return a value**.

1. **Function without parameters and without a return value**
2. **Function with parameters and without a return value**
3. **Function without parameters and with a return value**
4. **Function with parameters and with a return value**

### **1. Function Without Parameters and Without Return Value**

- The function **does not accept any parameters**.
- It **does not return any value**.
- It simply performs the specified business logic when called.

**Syntax:**

```python
def func_name():
    # Business Logic

func_name()
```

````python
def demo():
    print("This is a demo function.")
demo()
def main():
    print("This is the main function.")
main()
````

````python
# Create a function to print `N` natural numbers:

def nat_num():
    n=int(input("Enter the value of n"))
    for i in range(1,n+1):
        print(i)


nat_num()
````

````python
# Create a function to print `N` natural even numbers:

def even_num():
    n=int(input("Enter the value of n"))
    for i in range(1,n+1):
        if i%2==0:
            print(i)


even_num()
````

````python
# Create a function to print `N` natural palindrome numbers:

def pal_num():
    n=int(input("Enter the value of n"))
    for i in range(1,n+1):
        if str(i)==str(i)[::-1]:
            print(i)


pal_num()
````

### **1. Function With Parameters and Without Return Value**

- The function **does accept parameters**.
- It **does not return any value**.
- It simply performs the specified business logic when called.

**Syntax:**

```python
def func_name(var1,var2):
    # Business Logic

func_name(val1,val2)
```

````python
def demo(a,b,c):
    print(a)
    print(b)
    print(c)
demo(2,3,4)
````

````python
def details(name,age,address):
    print("Name",name)
    print("Age",age)
    print("Address",address)

details("Aakash",21,"Kmr")
````
