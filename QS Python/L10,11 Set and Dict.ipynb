## **Set**

* A **set** is a collection of **unique values**.
* It is mainly used when duplicate values need to be removed or when performing set operations.

**Syntax:**

```python
var = {val1, val2, val3, ..., valn}
```

### **Characteristics**

1. **Unordered** data type.
2. Does **not allow duplicate values**.
3. Does **not support indexing or slicing**.
4. It is **mutable**, but its elements must be **immutable/hashable**.
5. Empty set: `set()`

> **Note:** `{}` represents an **empty dictionary**, not an empty set. Use `set()` to create an empty set.

````python
#Empty Set
set1=set()
print(len(set1))
print(type(set1))
````

````python
set2={6,1,5,88,5,6,9,5,8,5,8,5,55,44,77,22,55,66,88,77,11}
set2
````

### **Some Built-in Set Methods**

1. **`add()`**

   * Adds a **single value** to the set.
   * Syntax: `var.add(value)`

2. **`update()`**

   * Adds **multiple values** to the set.
   * Syntax: `var.update({val1, val2, ..., valn})`

3. **`remove()`**

   * Removes a specified value from the set.
   * Raises a **`KeyError`** if the value does not exist.
   * Syntax: `var.remove(value)`

4. **`pop()`**

   * Removes and returns an **arbitrary element** from the set.
   * Syntax: `var.pop()`

5. **`clear()`**

   * Removes **all elements** from the set.
   * Syntax: `var.clear()`

![Mutable vs Immutable data](img.png)

````python
set2={1,2,3,6,5,4,2.2,3.3,55,False,"hii",(2,6,5,8)}
# Adding an element to the set
set2.add(100)
print(set2)

# Removing an element from the set
set2.remove(4)
print(set2)

# Updating the set with multiple elements
set2.update([1000,2000,3000])
print(set2)

# Popping an arbitrary element from the set
set2.pop()
print(set2)

# Clearing all elements from the set
set2.clear()
print(set2)
````

## **Dictionary**

* A **dictionary** is a collection of **key-value pairs**, where each key and its corresponding value together form an **item**.
* Each **key must be unique** within a dictionary.
* Dictionaries are useful for storing and accessing data using meaningful **keys**.

### **Syntax**

```python
var = {
    key1: val1,
    key2: val2,
    ...
    keyN: valN
}
```

### **Characteristics**

1. Stores data in **key-value pairs**.
2. **Keys must be unique and immutable (hashable)**, while values can be duplicated.
3. **Ordered** data type — preserves insertion order.
4. **Mutable** in nature.
5. Values are accessed using their **keys**, not index positions.
6. Does **not support indexing and slicing**.
7. Empty dictionary: `{}`

> **Note:** If duplicate keys are provided, the **latest value replaces the previous value**, while the key retains its original position.

**Example:**

```python
d1 = {'a': 1, 'b': 2, 'c': 3, 'a': 4}

print(d1)
# {'a': 4, 'b': 2, 'c': 3}
```

### **Dictionary Operations**

#### **1. Fetching a Value**

A value can be accessed from a dictionary using its corresponding **key**.

**Syntax:**

```python
var[key]
```

**Example:**

```python
student = {"name": "Aakash", "age": 22}

print(student["name"])  # Aakash
```

#### **2. Adding or Updating an Item**

The same syntax is used to **add a new key-value pair** or **update an existing value**.

**Syntax:**

```python
var[key] = value
```

**Example:**

```python
student["city"] = "Sultanpur"  # Add
student["age"] = 23             # Update
```

### **Some Built-in Dictionary Methods**

#### **1. `update()`**

* Adds new key-value pairs or **updates existing values**.
* **Syntax:** `var.update({key1: val1, key2: val2})`

#### **2. `pop()`**

* Removes the item associated with the **specified key** and returns its value.
* **Syntax:** `var.pop(key)`

#### **3. `clear()`**

* Removes **all key-value pairs** from the dictionary.
* **Syntax:** `var.clear()`

#### **4. `popitem()`**

* Removes and returns the **last inserted key-value pair** as a tuple.
* **Syntax:** `var.popitem()`

#### **5. `keys()`**

* Returns a view containing **all keys** of the dictionary.
* **Syntax:** `var.keys()`

#### **6. `values()`**

* Returns a view containing **all values** of the dictionary.
* **Syntax:** `var.values()`

#### **7. `items()`**

* Returns a view containing **all key-value pairs** as tuples.
* **Syntax:** `var.items()`


````python
# Dictionary
student = {
    "name": "Aakash",
    "age": 22,
    "branch": "Information Technology",
    "college": "KNIT Sultanpur",
    "cgpa": 8.5
}

print("Original Dictionary:")
print(student)

# 1. Accessing a value
print("\n1. Accessing a value:")
print(student["name"])

# 2. Adding a new item
student["city"] = "Sultanpur"
print("\n2. After adding an item:")
print(student)

# 3. Updating an existing value
student["cgpa"] = 8.8
print("\n3. After updating a value:")
print(student)

# 4. update() - Adding/updating multiple items
student.update({
    "age": 23,
    "skills": "Python, SQL, AI/ML"
})
print("\n4. After update():")
print(student)

# 5. keys() - Get all keys
print("\n5. keys():")
print(student.keys())

# 6. values() - Get all values
print("\n6. values():")
print(student.values())

# 7. items() - Get all key-value pairs
print("\n7. items():")
print(student.items())

# 8. pop() - Remove item using key
removed_value = student.pop("city")
print("\n8. After pop('city'):")
print(student)
print("Removed value:", removed_value)

# 9. popitem() - Remove the last inserted item
removed_item = student.popitem()
print("\n9. After popitem():")
print(student)
print("Removed item:", removed_item)

# 10. clear() - Remove all items
student.clear()
print("\n10. After clear():")
print(student)
````
