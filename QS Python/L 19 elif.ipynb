## **`if-elif` Statement**

* `elif` is a **keyword** that allows us to specify an additional condition.
* If the `if` condition is `False`, Python checks the `elif` conditions **one by one**.
* When a condition evaluates to `True`, its corresponding statement block is executed.
* We can use **multiple `elif` blocks** with a single `if` statement.
* It is useful when there are **multiple conditions with different statement blocks**.
* The `else` block is optional and executes when **all conditions are `False`**.
* Once a condition is satisfied, the remaining conditions are **not checked**.

**Syntax:**

```python
if condition:
    statement
elif condition:
    statement
elif condition:
    statement
else:
    statement
```

**Example:**

```python
num = int(input("Enter a number: "))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```
