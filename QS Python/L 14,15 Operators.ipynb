## **Operators**

* Operators are **symbols or keywords** used to perform specific operations on values.
* The values on which an operator performs an operation are called **operands**.

**Example:**

```python
2 + 3
```

* `+` → **Operator**
* `2, 3` → **Operands**
* `2 + 3` → **Operation**

> **In short:** Symbols/keywords → **Operators** | Values → **Operands**

### **Types of Operators**

| No. | Operator Type  | Operators                           | Output     |
| --- | -------------- | ----------------------------------- | ---------- |
| 1   | **Arithmetic** | `+`, `-`, `*`, `**`, `/`, `//`, `%` | Values     |
| 2   | **Relational** | `<`, `>`, `<=`, `>=`, `==`, `!=`    | Boolean    |
| 3   | **Assignment** | `=`                                 | Assignment |
| 4   | **Logical**    | `and`, `or`, `not`                  | Boolean    |
| 5   | **Bitwise**    | `&`, `\|`, `^`, `~`, `<<`, `>>`     | Values     |
| 6   | **Membership** | `in`, `not in`                      | Boolean    |
| 7   | **Identity**   | `is`, `is not`                      | Boolean    |

    
> **Note:** In Python, `|` is the **bitwise OR operator**. `||` is not a valid Python operator.

### **Arithmetic Operators**

* Arithmetic operators are used to perform **mathematical operations**.
* They generally return a **numeric value** as output.
* They are mainly used with numerical data types such as **`int`** and **`float`**.

### **Division Operators**

Python provides three types of division operations:

1. **`/` — True Division**

   * Returns the quotient in **decimal (float) form**.
   * Example: `10 / 3` → `3.3333...`

2. **`//` — Floor Division**

   * Returns the quotient after **flooring the result**.
   * Example: `10 // 3` → `3`

3. **`%` — Modulus**

   * Returns the **remainder** of the division.
   * Example: `10 % 3` → `1`

> **Note:** `//` does not simply remove the decimal part; it performs **floor division**, so the result is rounded down toward negative infinity.

> **Note:** `+` is used to concatenate two collections (`str`,`list` and `tuple`)

````python
a=2+3
b=6-3
c=5*9
d=8/2
e=3**3
f=8//2
g=8%3
print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
print(g)
````

## **Relational Operators**

* Relational operators are used to **compare two values** and determine their relationship.
* They return a **Boolean value**: `True` or `False`.

**Operators:**
`<`, `>`, `<=`, `>=`, `==`, `!=`

**Example:**

```python
a = 10
b = 20

print(a < b)   # True
print(a == b)  # False
```

### **Some Useful Functions**

#### **1. `ord()`**

* `ord('char')` returns the **Unicode code point** of a character.
* It is useful for **character comparison and string operations**.

**Syntax:**

```python
ord('A')  # 65
```

#### **2. `chr()`**

* `chr(value)` returns the **character corresponding to a Unicode code point**.

**Syntax:**

```python
chr(65)  # 'A'
```

> **Note:** In Python, `ord()` and `chr()` work with **Unicode**, not just ASCII.

### **Assignment Operator**

* The assignment operator `=` is used to **assign a value to a variable**.
* Arithmetic operators can be combined with `=` to form **compound assignment operators**.

**Examples:**

```python
x = 10
x += 5   # x = x + 5
x *= 2   # x = x * 2
```

### **Logical Operators**

Python provides three logical operators: **`and`**, **`or`**, and **`not`**.

* These are **keywords** used to combine or modify conditions.
* They are commonly used in **conditional and comparison expressions**.

| Operator | Purpose                                      |
| -------- | -------------------------------------------- |
| `and`    | True when **all conditions** are true        |
| `or`     | True when **at least one condition** is true |
| `not`    | **Reverses** the Boolean result              |

**Example:**

```python
age = 22
student = True

print(age > 18 and student)  # True
print(age < 18 or student)   # True
print(not student)           # False
```


## **Bitwise Operators**

* Bitwise operators perform operations on the **individual bits** of integer values.
* In Python, bitwise operators are primarily used with **integers**.
* The main bitwise operators are: `&`, `|`, `^`, `~`, `<<`, `>>`.

### **1. `&` — Bitwise AND**

* Compares corresponding bits of two integers.
* Returns `1` only when **both bits are `1`**.

**Example:**

```python
2 & 3
# 2 → 10
# 3 → 11
#     10 → 2
```

### **2. `|` — Bitwise OR**

* Compares corresponding bits of two integers.
* Returns `1` when **at least one bit is `1`**.

**Example:**

```python
2 | 3
# 2 → 10
# 3 → 11
#     11 → 3
```

### **3. `^` — Bitwise XOR**

* Compares corresponding bits of two integers.
* Returns `0` when both bits are **same** and `1` when they are **different**.

**Example:**

```python
5 ^ 5
# 101
# 101
# ---
# 000 → 0
```

### **4. `~` — Bitwise NOT**

* Inverts all the bits of an integer.
* Mathematically, Python evaluates it as:

```text
~n = -(n + 1)
```

**Example:**

```python
~-9
= -(-9 + 1)
= -(-8)
= 8
```

### **5. `<<` — Left Shift**

* Shifts the bits to the **left** by the specified number of positions.
* `n` zeros are added to the **right**.

**Example:**

```python
3 << 2
# 3 → 11
# Shift left by 2 → 1100
# 1100 → 12
```

### **6. `>>` — Right Shift**

* Shifts the bits to the **right** by the specified number of positions.
* The bits shifted out from the right are discarded.

**Example:**

```python
4 >> 2
# 4 → 100
# Shift right by 2 → 1
```

### **Binary Conversion Functions**

#### **`bin()`**

* Converts an integer into its **binary representation**.
* **Syntax:** `bin(int)`

```python
bin(10)  # '0b1010'
```

#### **`int()`**

* Can convert a binary string to its **integer value** by specifying base `2`.
* **Syntax:** `int(binary, 2)`

```python
int('1010', 2)  # 10
```

> **Note:** Python internally performs bitwise operations on the binary representation of integers and returns the result as an integer.
