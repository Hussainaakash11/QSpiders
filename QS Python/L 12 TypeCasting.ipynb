## **Type Casting**

* **Type casting** is the process of converting a value from **one data type to another**.
* Python provides built-in functions such as `int()`, `float()`, `str()`, `bool()`, etc. for type casting.

**Syntax:**

```python
new_var = Destination_DataType(var/value)
```

**Example:**

```python
x = "100"
y = int(x)

print(y)        # 100
print(type(y))  # <class 'int'>
```

### **Types of Type Casting**

* `int()` → Converts to integer
* `float()` → Converts to float
* `str()` → Converts to string
* `bool()` → Converts to Boolean


> ## **NOTE**

| Datatype | Main direct conversions |
|---|---|
| `int` | `float`, `str`, `bool`, `complex` |
| `float` | `int`, `str`, `bool`, `complex` |
| `str` | `int`, `float`, `bool`, `complex`, `list`, `tuple`, `set` |
| `bool` | `int`, `float`, `str`, `complex` |
| `complex` | `str`, `bool` |
| `list` | `str`, `bool`, `tuple`, `set`, `dict`* |
| `tuple` | `str`, `bool`, `list`, `set`, `dict`* |
| `set` | `str`, `bool`, `list`, `tuple`, `dict`* |
| `dict` | `str`, `bool`, `list`, `tuple`, `set`, `dict` |
| `None` | `str`, `bool` |

\* `dict()` requires an appropriate iterable of key-value pairs, such as `[("name", "Alex"), ("age", 20)]`.

```


````python
# Python automatically converts 'a' to int 
a = 7
print(type(a)) 

# Python automatically converts 'b' to float 
b = 3.0
print(type(b)) 

# Python automatically converts 'c' to float as it is a float addition 
c = a + b 
print(c) 
print(type(c))

# Python automatically converts 'd' to float as it is a float multiplication
d = a * b
print(d)
print(type(d))

# Example of type casting
a='2'
b='3'
x=int(a) + int(b)
print(x)
````
