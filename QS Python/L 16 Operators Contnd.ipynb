## **Membership Operators**

* Membership operators are used to **check whether a value is present in a collection or not**.
* They return a **Boolean value**: `True` or `False`.

### **Operators**

* **`in`** → Returns `True` if the value is present in the collection.
* **`not in`** → Returns `True` if the value is not present in the collection.

**Example:**

```python
name = "Python"

print("P" in name)       # True
print("z" not in name)   # True
```

````python
# Example
d1= {'a':10, 'b':20}
print(10 in d1)

print('2' in '235')

print(2 in [2,3,4])
````

## **Identity Operators**

* Identity operators are used to **check whether two operands refer to the same object in memory**.
* They compare the **identity (memory reference)** of two objects.
* They return a **Boolean value**: `True` or `False`.

### **Operators**

* **`is`** → Returns `True` if both operands refer to the **same object**.
* **`is not`** → Returns `True` if both operands refer to **different objects**.

**Example:**

```python
a = [10, 20]
b = a
c = [10, 20]

print(a is b)      # True
print(a is c)      # False
print(a is not c)  # True
```
