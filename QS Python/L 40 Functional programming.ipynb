### **1. Anonymous Function / Lambda Function:**
- This is a type of function that has no name.
- In Python, `lambda` is a keyword used to define an anonymous function / lambda expression.
- It is defined using a single expression, hence it is also known as a single-line function.
- It can take `n` number of arguments (inputs), but evaluates to a single output.
- This function always returns the evaluated value automatically (without using the `return` keyword).
- It is used to perform simple calculations.
- `if-else` conditions can be implemented using an anonymous function (via ternary expressions).
- Syntax:
```python
var = lambda arguments:expression/operation
print(var(args))

# usinf if-else

var= lambda arguments : TSB if condition else FSB
```


````python
# WAP to get square of a number using lambda function.
square = lambda x: x ** 2
print(square(5))
````

````python
# WAP which returns sum of two numbers using lambda function.
sum = lambda a, b: a + b
print(sum(5, 10))
````

````python
# Find the cube of a number using lambda function.
cube = lambda x: x ** 3
print(cube(3))
````

````python
# WAP which takes first name and last name as input and returns full name using lambda function.
full_name = lambda first_name, last_name: first_name + " " + last_name
print(full_name("Aakash", "Hussain"))
````

````python
odd_even = lambda x: "Even" if x % 2 == 0 else "Odd"
print(odd_even(5))
print(odd_even(10))
````

````python
odd_even = lambda x:x % 2 == 0
print(odd_even(5))
print(odd_even(10))
````

````python
is_odd = lambda x: x % 2 != 0
print(is_odd(5))

is_even = lambda x: x % 2 == 0
print(is_even(10))

is_palindrome = lambda s: s == s[::-1]
print(is_palindrome("madam"))

is_float = lambda x: type(x) == float
print(is_float(10.5))
````

### **2. Map Function:**
- It is used to traverse through an iterable and apply the same functionality to each and every element.
- It is used to process and modify a collection.
- The length of the input is the same as the length of the output.
- `map()` takes two primary arguments:
    1. Function (name or lambda expression)
    2. Iterable (one or more)
- The function passed to `map()` must return a value.
- map() always returns the map object which is iterable. So to get value we can use the loop over the map object or we can typecast it.
- Syntax:
```python 
var = map(function_name, iterable)
print(var)
```
- Function can be Inbuit, userdefined or anonymous function.

````python
l1 = [1, 2, 3, 4, 5]

def pro_ten(x):
    return x * 10

l2 = list(map(pro_ten, l1))
print(l2)
````

````python
l1 = [1, 2, 3, 4, 5]

pro_ten = lambda x: x * 10
l2 = list(map(pro_ten, l1))
print(l2)
````

````python
l1 = [1, 2, 3, 4, 5]

l2 = list(map(lambda x: x * 10, l1))
print(l2)
````

````python
s = "She is beautiful"
# Output: [3, 2, 9]
l2 = list(map(len, s.split()))
print(l2)
````

````python
s="abc"
l2 = list(map(ord, s))
print(l2)
````

````python
t1 = (1, 2, 3, 4, 5)
l2 = list(map(lambda x: x **3 if x%2 !=0 else x**2, t1))
print(l2)
````

````python
n = 5
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

l2 = list(map(factorial, range(1,n+1)))
print(l2)
````

### **3. Filter Function:**
- It is an inbuilt function which is used to traverse through the collection and filter some specific values.
- It is used to process and filter a collection.
- Length of output may be equal or less than length of the input.
- `filter()` takes two primary arguments:
    1. Function (name or lambda expression)
    2. Iterable (one or more)
- The function passed to `filter()` must return a boolean value.
- The filter always returns the filter object which is iterable.
- So to get the value we can use the loop over the map object or we can type cast it.


````python
l1 = [1, 2,'bye', 3, 4, 'hii', 5]
l2= list(filter(lambda x: type(x) == int, l1))
print(l2)
````

````python
# Filter all the palindrome numbers:
l1 = [121, 123, 131, 141, 151]
l2 = list(filter(lambda x: str(x) == str(x)[::-1], l1))
print(l2)
````

````python
# Find all the prime numbers from a given range.
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

range_start = 10
range_end = 50
prime_numbers = list(filter(is_prime, range(range_start, range_end + 1)))
print(prime_numbers)
````

````python
l1 = [2,'python',4,5,'java']
# op [2,24,120]
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
l2 = list(map(factorial,filter(lambda x: type(x) == int, l1)))
print(l2)
````

````python
t1 = (1, 2, 3, 4, 5,6)
# [200, 400, 600]
l2 = list(map(lambda x: x * 100, filter(lambda x: x % 2 == 0, t1)))
print(l2)
````

### **4. Reduce Function:**

- It is a standard library function used to reduce an iterable into a single cumulative value.
- It uses a mathematical folding technique.
- It is present in the `functools` module.
- It takes two required arguments (and an optional third `initializer` argument):
  1. Function (name or lambda expression)
  2. Iterable
- The function used in `reduce()` must accept two parameters and return a single value.
- `reduce()` always returns a single value, so we do not need to typecast it.

````python
# Sum of values in a list using reduce function.
from functools import reduce
l1 = [1, 2, 3, 4, 5]
sum_of_values = reduce(lambda x, y: x + y, l1)
print(sum_of_values)
````

````python
# Product of values in a list using reduce function.
from functools import reduce
l1 = [1, 2, 3, 4, 5]
product_of_values = reduce(lambda x, y: x * y, l1)
print(product_of_values)
````

````python
# Find the factorial of a number using reduce function.
from functools import reduce
n = 5
factorial_of_n = reduce(lambda x, y: x * y, range(1, n + 1))
print(factorial_of_n)
````

````python
# Find the smallest number in a list using reduce function.
from functools import reduce
l1 = [5, 2, 9, 1, 7]
smallest_number = reduce(lambda x, y: x if x < y else y, l1)
print(smallest_number)
````

````python
# Find the sum of factorial of all the numbers in a list using reduce function.
from functools import reduce
l1 = [2,4,"python","java",1]

# def factorial(n):
#     if n == 0:
#         return 1
#     else:
#         return n * factorial(n-1)
fact = lambda n: 1 if n == 0 else n * fact(n - 1)
l2 = reduce(lambda x, y: x + y, map(fact, filter(lambda x: type(x) == int, l1)))
print(l2)
````
