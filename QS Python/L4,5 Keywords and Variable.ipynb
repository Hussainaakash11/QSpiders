## What are Keywords in Python?

**Definition:** Keywords are reserved words in Python that have special meaning and specific purposes. They are part of the Python syntax and cannot be used as variable names, function names, or class names.

### Characteristics of Keywords:
1. **Reserved** - Cannot be used as identifiers (variable/function/class names)
2. **Case-Sensitive** - `True` is a keyword, but `true` is not
3. **Pre-defined** - Built into the Python language
4. **Fixed Number** - Python 3.x has 35 keywords
5. **Syntax-Based** - Used to define the structure and logic of programs

### Common Python Keywords (35 total):

**Control Flow:** if, elif, else, for, while, break, continue, pass
**Functions:** def, return, lambda, yield
**Classes & OOP:** class, super, with, as
**Error Handling:** try, except, finally, raise
**Import:** import, from, as
**Boolean:** True, False, and, or, not
**Other:** None, in, is, del, global, nonlocal, assert

### Examples of Keyword Usage:
- `if x > 5:` - control flow
- `def my_function():` - function definition
- `for i in range(10):` - iteration
- `class MyClass:` - class definition
- `try: ... except:` - error handling
- `import numpy` - importing modules

## Boolean Values in Python

**Definition:** Boolean is a data type that can only have one of two values: `True` or `False`. It's used for conditional statements and logical operations.

### Special Boolean Constants:

**1. True**
   - Represents logical true
   - Equivalent to integer value: **1**
   - Used in conditions: `if True:` (always executes)
   - Example: `x = True` creates a boolean variable

**2. False**
   - Represents logical false
   - Equivalent to integer value: **0**
   - Used in conditions: `if False:` (never executes)
   - Example: `y = False` creates a boolean variable

**3. None**
   - Represents absence of value or null
   - Neither True nor False
   - Default return value of functions without explicit return
   - Used to indicate "nothing" or "no value"

### Boolean Arithmetic (True = 1, False = 0):

**Addition:**
- `True + True` = 1 + 1 = **2**
- `True + False` = 1 + 0 = **1**
- `False + False` = 0 + 0 = **0**

**Multiplication:**
- `True * True` = 1 × 1 = **1**
- `True * False` = 1 × 0 = **0**
- `False * False` = 0 × 0 = **0**

**Subtraction & Division:**
- `True - False` = 1 - 0 = **1**
- `True / True` = 1 ÷ 1 = **1.0**

### Boolean Operators:

**and** - Returns True if both conditions are True
- `True and True` → True
- `True and False` → False
- `False and False` → False

**or** - Returns True if at least one condition is True
- `True or False` → True
- `False or False` → False
- `True or True` → True

**not** - Reverses the boolean value
- `not True` → False
- `not False` → True

### Use Cases:
- Conditional statements: `if is_valid:`
- Loop conditions: `while is_running:`
- Storing true/false states: `is_logged_in = True`
- Logical operations in complex conditions

### What are Variables?

**Definition:** A variable is a named container or placeholder used to store data values in a program's memory. It acts as a reference to a value that can be used and modified throughout the program.

### Variable Characteristics:

1. **Storage Containers**
   - Hold data of any type (integers, strings, floats, lists, etc.)
   - Memory is allocated when a variable is created
   - The value can be accessed using the variable name

2. **Naming Convention**
   - Use descriptive names: `student_age`, `total_price`, `is_valid`
   - Start with letter or underscore: `_name`, `name` (not `9name`)
   - Use lowercase with underscores (snake_case): `my_variable`
   - Avoid Python keywords as variable names
   - Case-sensitive: `age` and `Age` are different variables

3. **Declaration & Assignment**
   - Syntax: `variable_name = value`
   - Python is dynamically typed: type inferred from value
   - No need to declare type explicitly
   - Example: `x = 10` (integer), then `x = "hello"` (string)

### Data Types Variables Can Hold:

| Type | Example | Explanation |
|------|---------|-------------|
| **int** | `age = 25` | Whole numbers (positive, negative, zero) |
| **float** | `price = 19.99` | Decimal numbers |
| **str** | `name = "John"` | Text/string data |
| **bool** | `is_valid = True` | True or False values |
| **list** | `nums = [1, 2, 3]` | Ordered collection of items |
| **dict** | `user = {"name": "John"}` | Key-value pairs |
| **tuple** | `point = (5, 10)` | Immutable ordered collection |

### Variable Scope:

**Local Variables**
- Defined inside functions
- Accessible only within that function
- Deleted when function ends

**Global Variables**
- Defined outside functions
- Accessible throughout the program
- Can be modified using `global` keyword

**Multiple Variable Creator**
- `Var1, Var2, Var3 = Val1, Val2, Val3`

### Best Practices:
✓ Use meaningful, descriptive names
✓ Keep names concise but clear
✓ Use consistent naming convention (snake_case)
✓ Avoid single-letter names (except loop counters)
✓ Don't reuse variable names for different purposes
✓ Initialize variables before using them

### Memory Management in Python

**Understanding Python's Memory Model:**

Python uses a two-part memory system to store data efficiently:

#### **1. Stack Memory**
- Stores variable **names/references** (lightweight)
- Stores small, fixed-size data (integers, booleans)
- Memory is automatically freed when variables go out of scope
- LIFO (Last In, First Out) structure
- Faster access

**Example:**
```python
a = 10  # Variable 'a' stored in Stack
b = 20  # Variable 'b' stored in Stack
```

#### **2. Heap Memory**
- Stores actual **values** and objects
- Stores large, variable-size data (strings, lists, dictionaries)
- Memory is managed by garbage collector
- Less organized, slower access than Stack
- Larger capacity than Stack

**Example:**
```python
name = "John"  # Variable 'name' in Stack, actual string "John" in Heap
my_list = [1, 2, 3]  # Variable 'my_list' in Stack, actual list in Heap
```

---

#### **3. Memory Address (ID)**

**What is an ID?**
- Every value stored in memory has a unique memory address
- The `id()` function returns this address (integer) where the value is stored
- Different variables can reference the same value (same address)

**Syntax:**
```python
id(variable_name)  # Returns the memory address as an integer
```

**Example:**
```python
x = 5
print(id(x))  # Output: 140734819892616 (example address)

y = 5
print(id(y))  # Output: 140734819892616 (same address - Python optimizes small integers)

z = 10
print(id(z))  # Output: 140734819892648 (different address)
```

**Key Points:**
- `id()` always returns an integer (memory address)
- Same values can share the same address (optimization for immutable objects)
- Different values have different addresses
- IDs change between program runs (addresses are randomized for security)

#### **Memory Diagram Example:**

```

┌─────────────┐                 ┌──────────┐
│ a ─────────────────────────> │ 10       │
└─────────────┘                 └──────────┘
┌─────────────┐                 ┌──────────┐

┌─────────────┐                 ┌──────────┐

└─────────────┘                 └──────────┘
```

#### **Why This Matters:**
1. **Performance:** Understanding memory helps optimize code
2. **Mutable vs Immutable:** Affects how variables behave
3. **Memory Leaks:** Helps prevent unused memory accumulation
4. **Object Identity:** `id()` is used to check if two variables reference the same object
5. **Debugging:** Memory addresses help trace object creation and lifetime

#### **Identity vs Equality:**

```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)      # True (same content)
print(a is b)      # False (different objects, different addresses)
print(id(a) == id(b))  # False (different memory addresses)

c = a
print(c is a)      # True (same object, same address)
print(id(c) == id(a))  # True (same memory address)
```

## **Dynamic Memory Allocation**

Dynamic Memory Allocation is the process of allocating and managing memory during the **runtime** of a program.

### Reference Counting

**Reference Counting** is a memory-management technique that keeps track of how many variables or references point to a particular value/object.

> **Reference Count = Number of variables/references pointing to an object**

* When a variable is assigned to an object, its **reference count increases**.
* When a reference is removed, its **reference count decreases**.
* When the reference count becomes **zero**, the object is no longer accessible.

### Example

```text
A → Object ← B
```

Here, the object's **reference count = 2** because two variables (`A` and `B`) refer to it.

---

### Garbage Collector

A **Garbage Collector (GC)** automatically identifies and removes objects from memory that are **no longer referenced or accessible** by the program.

### Key Points

* It removes **unused/unreferenced objects** from memory.
* Objects that are no longer assigned to or reachable by any variable can be collected.
* Garbage collection happens **automatically** in languages that support automatic memory management.
* Its main purpose is to **free memory and prevent unnecessary memory consumption**.

### Example

```text
A → Object
```

If `A` is removed:

```text
A ✕    Object
       ↓
   Unreferenced
```

The object becomes eligible for **Garbage Collection**.

> **In short:** Reference Counting tracks how many references point to an object, while the Garbage Collector frees memory occupied by objects that are no longer needed.

## 🔄 **Swapping Values in Python**

**Swapping** means exchanging the values stored in two variables.

Python provides multiple ways to swap values. The two common approaches are:

---

### 1. Using a Temporary Variable

A third variable, called `temp`, is used to temporarily store one of the values.

```python
a = 10
b = 20

temp = a
a = b
b = temp
```

### Step-by-step

```text
Initial:
a = 10
b = 20

temp = a   → temp = 10
a = b      → a = 20
b = temp   → b = 10
```

**Final values:**

```text
a = 20
b = 10
```

---

### 2. Using Multiple Assignment

Python allows us to swap two variables directly using **multiple assignment**. No temporary variable is required.

```python
a = 299
b = 898

a, b = b, a
```

### Step-by-step

```text
Before:
a = 299
b = 898

After:
a = 898
b = 299
```

### ⭐ Recommended Pythonic Approach

```python
a, b = b, a
```

This is the **preferred and most concise way** to swap values in Python.

> **Note:** This is better described as **multiple assignment / tuple unpacking**.

### Quick Comparison

| Method              | Temporary Variable | Code          |
| ------------------- | ------------------ | ------------- |
| Using `temp`        | ✅ Required         | `temp = a`    |
| Multiple Assignment | ❌ Not required     | `a, b = b, a` |

**Key Takeaway:**

> In Python, use `a, b = b, a` for a clean, readable, and idiomatic value swap.
