## Q2: Explain what **dynamically typed language** means.

**Answer:**
A dynamically typed language determines variable types at **runtime** (when code executes), not at compile-time.

**Key Points:**
- No need to declare variable types explicitly
- Type is inferred from the assigned value
- A variable can change types during execution
- Type errors are caught when that line runs, not before

**Example:**
```python
x = 5           # x is integer
x = "hello"     # x is now string (type changed)
x = [1, 2, 3]   # x is now list (type changed again)
```

**Advantages:**
✓ Faster coding (less boilerplate)
✓ More flexible
✓ Easier for beginners

**Disadvantages:**
✗ Errors only caught at runtime
✗ Slightly slower execution
✗ Requires thorough testing

**Contrast:** Statically typed languages (C, Java) require type declaration: `int x = 5;`

## Q3: Difference between **interpreter and compiler**.

**Answer:**

| Aspect | Compiler | Interpreter |
|--------|----------|-------------|
| **Translation** | Translates entire code at once to machine code | Translates and executes line-by-line |
| **Output** | Produces executable file (.exe, .o) | No executable file; runs directly |
| **Execution Speed** | Very fast (pre-compiled) | Slower (real-time translation) |
| **Development** | Slow (must compile each time) | Fast (immediate testing) |
| **Error Detection** | Before execution (compile-time) | During execution (runtime) |
| **Platform** | Platform-specific (must recompile for each OS) | Platform-independent (needs interpreter) |
| **Source Code** | Can be hidden in compiled binary | Source code must be provided |
| **Debugging** | Harder (need debugging tools) | Easier (see errors immediately) |
| **Examples** | C, C++, Rust, Go | Python, JavaScript, Ruby, PHP |

## Q4: What compiles the code in Python?

**Answer:**
Python code is compiled by the **Python compiler** (written in C), which converts source code into **bytecode** (.pyc files).

**The Process:**
1. Python compiler translates `.py` files → `.pyc` files (bytecode)
2. Bytecode is stored in `__pycache__` directory
3. **Python Virtual Machine (PVM)** executes the bytecode
4. This happens automatically; users don't directly control it

**Key Point:** Python uses a hybrid approach:
- **Compilation:** Source code → Bytecode
- **Interpretation:** Bytecode → Machine code execution (by PVM)

## Q5: Is Python **interpreted or compiled**? Explain why.

**Answer:**
Python is **both interpreted AND compiled**, using a hybrid approach:

**Technically, Python is "Interpreted"** because:
- Users don't explicitly compile code
- Code executes without creating standalone executables
- Errors found during execution, not before
- Requires Python interpreter to run

**But internally, Python is "Compiled"** because:
- Python compiles source code to bytecode (.pyc) before execution
- This bytecode is what the Python Virtual Machine (PVM) actually runs
- Bytecode is platform-independent, improving performance

**Why This Hybrid Approach?**
- **Faster execution than pure interpretation** (bytecode is pre-compiled)
- **Platform independence** (bytecode runs on any OS with Python)
- **User convenience** (no manual compilation step)
- **Combines best of both worlds**

## Q7: What are **keywords in Python**?

**Answer:**
Keywords are reserved words that have special meaning in Python syntax. They cannot be used as variable, function, or class names.

**Characteristics:**
- Reserved for specific purposes
- Case-sensitive (`True` ≠ `true`)
- Cannot be modified or overwritten
- Python 3.x has **35 keywords**

**Common Keywords:**
- **Control Flow:** if, elif, else, for, while, break, continue, pass
- **Functions:** def, return, lambda, yield
- **Classes:** class, super, with, as
- **Error Handling:** try, except, finally, raise
- **Imports:** import, from, as
- **Boolean:** True, False, and, or, not
- **Others:** None, in, is, del, global, nonlocal, assert

**Example:**
```python
if x > 5:          # if, > are keywords/operators
    print(x)       # print is a function, not a keyword
    
def my_func():     # def is a keyword
    return x       # return is a keyword
```

## Q8: How to fetch all the **keywords**?

**Answer:**
Use the **`keyword` module** to fetch all keywords programmatically:

**Method 1: Using `keyword.kwlist`**
```python
import keyword
print(keyword.kwlist)
```

**Output:**
```
['False', 'None', 'True', 'and', 'as', 'assert', 'async', 'await', 'break', 
'class', 'continue', 'def', 'del', 'elif', 'else', 'except', 'finally', 'for', 
'from', 'global', 'if', 'import', 'in', 'is', 'lambda', 'nonlocal', 'not', 'or', 
'pass', 'raise', 'return', 'try', 'while', 'with', 'yield']
```

**Method 2: Count total keywords**
```python
import keyword
print(f"Total keywords: {len(keyword.kwlist)}")
# Output: Total keywords: 35
```

**Method 3: Check if a word is a keyword**
```python
import keyword
print(keyword.iskeyword('if'))      # True
print(keyword.iskeyword('ifx'))     # False
print(keyword.iskeyword('print'))   # False (print is a function)
```

**Method 4: Print formatted list**
```python
import keyword
for i, kw in enumerate(keyword.kwlist, 1):
    print(f"{i}. {kw}")
```

**Why This Matters:**
- Helps avoid naming conflicts
- Useful for code validation
- Educational to understand language structure
