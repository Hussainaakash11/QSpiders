## **Control Statements**

* By default, Python executes statements **from top to bottom** and expressions generally **from left to right**.
* **Control statements** are used to control or alter the normal flow of program execution.
* There are two main types of control statements:

1. **Conditional / Decision-Making Statements**

   * Used to execute code based on a **condition**.

2. **Looping Statements**

   * Used to **repeat a block of code** multiple times.

## **Conditional Statements**

* Conditional statements control the **flow of execution based on conditions**.
* They are used to make **decisions** in a program.

### **Types of Conditional Statements**

1. `if`
2. `if-else`
3. `if-elif`
4. **Nested `if`**

## **`if` Statement**

* `if` is a **keyword** used to specify a condition.
* The statement block inside `if` executes **only when the condition is `True`**.
* Python uses **indentation** to define the statement block.

**Syntax:**

```python
if condition:
    statement
    statement
    ...
```

**Example:**

```python
age = 20

if age >= 18:
    print("Eligible to vote")
```

````python
num=int(input("Enter Num"))
if num>0:
    print("Num is positive")
````

## **`else` Statement**

* `else` is a **keyword** that provides an alternative block for an `if` statement.
* It does **not contain a condition**.
* The `else` block executes when the `if` condition evaluates to **`False`**.
* A single `if` statement can have **only one `else` block**.

**Syntax:**

```python
if condition:
    statement
else:
    statement
```

**Example:**

```python
num = int(input("Enter a number: "))

if num > 0:
    print("Number is positive")
else:
    print("Number is negative")
```
