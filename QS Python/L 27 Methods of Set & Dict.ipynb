# Python `set` and `dict` Methods

## 1. Set Methods

### Modification Methods
| Method | Purpose | Syntax |
|---|---|---|
| `add()` | Adds a single element | `s.add(value)` |
| `update()` | Adds multiple elements | `s.update(iterable)` |
| `remove()` | Removes a specified element; raises `KeyError` if absent | `s.remove(value)` |
| `discard()` | Removes a specified element; no error if absent | `s.discard(value)` |
| `pop()` | Removes and returns an arbitrary element | `s.pop()` |
| `clear()` | Removes all elements | `s.clear()` |

### Set Operation Methods
| Method | Purpose | Syntax |
|---|---|---|
| `union()` | Combines elements from sets | `s1.union(s2)` |
| `intersection()` | Returns common elements | `s1.intersection(s2)` |
| `difference()` | Returns elements only in first set | `s1.difference(s2)` |
| `symmetric_difference()` | Returns elements present in either set, but not both | `s1.symmetric_difference(s2)` |

### Update Set Methods
These modify the existing set.

| Method | Purpose |
|---|---|
| `intersection_update()` | Keeps only common elements |
| `difference_update()` | Removes elements found in another set |
| `symmetric_difference_update()` | Keeps elements present in either set, but not both |

### Relationship / Comparison Methods
| Method | Purpose |
|---|---|
| `isdisjoint()` | Checks whether two sets have no common elements |
| `issubset()` | Checks whether one set is a subset of another |
| `issuperset()` | Checks whether one set is a superset of another |

### Copy
```python
s.copy()
```
Creates a **shallow copy** of the set.

### Set Operators

```python
s1 | s2   # Union
s1 & s2   # Intersection
s1 - s2   # Difference
s1 ^ s2   # Symmetric difference
s1 <= s2  # Subset
s1 >= s2  # Superset
```

---

### Quick Memory Trick

**Set:**  
`add → update → remove/discard → pop → clear`  
`union → intersection → difference → symmetric_difference`


# 2. Dictionary Methods

| Method | Purpose | Syntax |
|---|---|---|
| `clear()` | Removes all items | `d.clear()` |
| `copy()` | Creates a shallow copy | `d.copy()` |
| `fromkeys()` | Creates dictionary from given keys | `dict.fromkeys(keys)` |
| `get()` | Returns value for a key | `d.get(key)` |
| `items()` | Returns key-value pairs | `d.items()` |
| `keys()` | Returns all keys | `d.keys()` |
| `pop()` | Removes specified key and returns its value | `d.pop(key)` |
| `popitem()` | Removes and returns the last inserted item | `d.popitem()` |
| `setdefault()` | Returns value of key; adds key if absent | `d.setdefault(key, value)` |
| `update()` | Adds/updates key-value pairs | `d.update(other)` |
| `values()` | Returns all values | `d.values()` |

### Important Examples

```python
d = {"a": 10, "b": 20}

d.get("a")              # 10
d.keys()                # dict_keys(['a', 'b'])
d.values()              # dict_values([10, 20])
d.items()               # dict_items([('a', 10), ('b', 20)])

d.update({"c": 30})     # Add/update
d.setdefault("d", 40)   # Add if key doesn't exist

d.pop("a")              # Remove 'a'
d.popitem()             # Remove last item

d.copy()                # Shallow copy
d.clear()               # Remove everything
```

### Quick Memory Trick

**Dictionary:**  
`get → keys → values → items → update → pop → popitem → setdefault → clear → copy → fromkeys`
