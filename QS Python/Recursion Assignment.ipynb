## 1. Print the digits of a number

```python
def print_digits(n):
    if n == 0:
        return
    print_digits(n // 10)
    print(n % 10)

num = int(input("Enter a number: "))
print_digits(num)
```

## 2. Print the odd digits of a number

```python
def print_odd_digits(n):
    if n == 0:
        return
    print_odd_digits(n // 10)
    digit = n % 10
    if digit % 2 != 0:
        print(digit)

num = int(input("Enter a number: "))
print_odd_digits(num)
```

## 3. Print the even digits of a number

```python
def print_even_digits(n):
    if n == 0:
        return
    print_even_digits(n // 10)
    digit = n % 10
    if digit % 2 == 0:
        print(digit)

num = int(input("Enter a number: "))
print_even_digits(num)
```

## 4. Print the digits multiple of five in a given number

```python
def print_digits_multiple_of_five(n):
    if n == 0:
        return
    print_digits_multiple_of_five(n // 10)
    digit = n % 10
    if digit % 5 == 0:
        print(digit)

num = int(input("Enter a number: "))
print_digits_multiple_of_five(num)
```

## 5. Reverse a number

```python
def reverse_number(n, reverse=0):
    if n == 0:
        return reverse
    digit = n % 10
    reverse = reverse * 10 + digit
    return reverse_number(n // 10, reverse)

num = int(input("Enter a number: "))
print("Reverse =", reverse_number(num))
```

## 6. Print the sum of n natural numbers

```python
def natural_sum(n):
    if n == 0:
        return 0
    return n + natural_sum(n - 1)

n = int(input("Enter n: "))
print("Sum =", natural_sum(n))
```

## 7. Print the factorial of a number (product of n natural numbers)

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

n = int(input("Enter number: "))
print("Factorial =", factorial(n))
```

## 8. Sum of digits of a number

```python
def sum_of_digits(n):
    if n == 0:
        return 0
    return (n % 10) + sum_of_digits(n // 10)

num = int(input("Enter a number: "))
print("Sum of digits =", sum_of_digits(num))
```

## 9. Product of digits of a number

```python
def product_of_digits(n):
    if n < 10:
        return n
    return (n % 10) * product_of_digits(n // 10)

num = int(input("Enter a number: "))
print("Product of digits =", product_of_digits(num))
```

## 10. Find the sum of values in a list

```python
def list_sum(values, index=0):
    if index == len(values):
        return 0
    return values[index] + list_sum(values, index + 1)

numbers = [10, 20, 30, 40, 50]
print("Sum =", list_sum(numbers))
```
