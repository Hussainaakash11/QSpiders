## **List**

* A **list** is a collection that can store both **homogeneous and heterogeneous data**.
* It can store elements of **different data types** in a single variable.

**Syntax:**

```python
var = [val1, val2, val3, ..., valn]
```

### **Characteristics**

* **Ordered** data type.
* Allows **duplicate values**.
* **Mutable** data type.
* Supports **indexing and slicing**.
* Can store **multiple data types**.
* Empty list: `[]`

### **Modifying a Value**

A list element can be modified using its **index position**.

**Syntax:**

```python
var[index] = newValue
```

````python
# Mutablity
# To Modify a Value: `var[index]=newValue`

l1=[2,5,8,"hello",2.5,True]
print(l1)

l1[2]=68
print(l1)
````

````python
#Nested List
l3=[2,6,8,44,4.5,['a','v','c'],True]
print(l3)

#Indexing in Nested List
ans=l3[5][1]
print(ans)
````

````python
#Indexing
l3=[2,6,8,44,4.5,['a','v','c'],True]
ans=l3[5][1:-1]
print(ans)

# Type Function inside List
print(type(ans))
````

### **Some List Methods**

1. **`append()`**

   * Adds a value to the **end** of the list.
   * Syntax: `var.append(value)`

2. **`extend()`**

   * Adds **multiple values** to the end of the list.
   * Syntax: `var.extend([val1, val2, ..., valn])`

3. **`insert()`**

   * Adds a value at a **specific index**.
   * Syntax: `var.insert(index, value)`

4. **`remove()`**

   * Removes the **first occurrence** of a specified value.
   * Syntax: `var.remove(value)`

5. **`pop()`**

   * Removes and **returns** the value at a specified index.
   * By default, it removes the **last value**.
   * Syntax: `var.pop(index)`

6. **`clear()`**

   * Removes **all elements** from the list.
   * Syntax: `var.clear()`

## **Tuple**

* A **tuple** is a collection that can store both **homogeneous and heterogeneous data**, similar to a list.
* A tuple is an **immutable collection**, meaning its elements cannot be modified after creation.
* It is useful for storing data that should remain **unchanged**.

**Syntax:**

```python
var = (val1, val2, ..., valn)
```

### **Characteristics**

* **Ordered** data type.
* Allows **duplicate values**.
* **Immutable** data type.
* Supports **indexing and slicing**.
* Empty tuple: `()`

> **Note:** To create a tuple with a **single element**, a comma must be added after the element.

```python
t = (10,)   # Tuple
x = (10)    # Integer
```
