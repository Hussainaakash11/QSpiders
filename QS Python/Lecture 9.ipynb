## **List**
* List is a collection of homogeneous and heterogeneous data.
* List is a data type which can store all the other data types.
* *Syntax:* `var =[val1, val2, val3,---,valn]`.

### **Characteristics:**
* Ordered datatype
* Allows duplicate values
* Mutable data type
* Supports indexing and slicing
* `EmptyList -> []`
* *To Modify a value:* `var[index]=newValue`

```python
# Mutablity
# To Modify a Value: `var[index]=newValue`

l1=[2,5,8,"hello",2.5,True]
print(l1)

l1[2]=68
print(l1)
```

```python
#Nested List
l3=[2,6,8,44,4.5,['a','v','c'],True]
print(l3)

#Indexing in Nested List
ans=l3[5][1]
print(ans)
```

```python
#Indexing
l3=[2,6,8,44,4.5,['a','v','c'],True]
ans=l3[5][1:-1]
print(ans)

# Type Function inside List
print(type(ans))
```

### **Some Functions of List**

1. `append()`:
    * to add value at end: `var.append(val)`
2. `extend()`:
    * to add more than one value at end: `var.extend([val1,val2,---valn])`
3. `insert()`:
    * to add value at any particular index: `var.insert(index,value)`
4. `remove()`:
    * to remove any particular value: `var.remove(value)`
5. `pop()`:
    * to remove from particular index.
    * It returns the removed value.
    * By default is removes last value.
    * Syntax: `var.pop(index)`
6. `clear()`:
    * to clear all the values from the list: `var.clear()`
