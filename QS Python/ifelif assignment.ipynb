## 1. Age Category

```python
age = int(input("Enter your age: "))
if age < 13:
    print("Child")
elif age <= 19:
    print("Teenager")
elif age <= 64:
    print("Adult")
else:
    print("Senior")
```

## 2. Grade Assignment

```python
score = float(input("Enter your score (0-100): "))
if score < 0 or score > 100:
    print("Invalid score")
elif score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("F")
```

## 3. Positive, Zero, or Negative

```python
num = float(input("Enter a number: "))
if num > 0:
    print("Positive")
elif num == 0:
    print("Zero")
else:
    print("Negative")
```

## 4. FIZZ, BUZZ, FIZZBUZZ

```python
num = int(input("Enter a number: "))
if num % 3 == 0 and num % 5 == 0:
    print("FIZZBUZZ")
elif num % 3 == 0:
    print("FIZZ")
elif num % 5 == 0:
    print("BUZZ")
```

## 5. Number of Digits

```python
num = abs(int(input("Enter a number: ")))
if num <= 9:
    print("Single digit")
elif num <= 99:
    print("Two digit")
elif num <= 999:
    print("Three digit")
else:
    print("More than three digit")
```

## 6. Electricity Bill

```python
units = float(input("Enter units consumed: "))

if units < 0:
    print("Invalid units")
elif units <= 100:
    bill = units * 5
    print(bill)
elif units <= 300:
    bill = (100 * 5) + (units - 100) * 7.5
    print(bill)
elif units <= 500:
    bill = (100 * 5) + (200 * 7.5) + (units - 300) * 10
    print(bill)
else:
    bill = (100 * 5) + (200 * 7.5) + (200 * 10) + (units - 500) * 15
    print(bill)
```

```python
# Failed Attempt
units = float(input("Enter units consumed: "))
if units < 0:
    print("Invalid units")
elif units <= 100:
    print("Electricity bill = ₹", units * 5)
elif units <= 300:
    print("Electricity bill = ₹", units * 7.5)
elif units <= 500:
    print("Electricity bill = ₹", units * 10)
else:
    print("Electricity bill = ₹", units * 15)
```

## 7. Greeting by Hour

```python
hour = int(input("Enter current hour (0-23): "))
if hour < 0 or hour > 23:
    print("Invalid hour")
elif hour <= 11 and hour >= 5:
    print("Good Morning")
elif hour <= 16:
    print("Good Afternoon")
elif hour <= 20:
    print("Good Evening")
else:
    print("Good Night")
```

## 8. Traffic Light Decision

```python
color = input("Enter traffic light color: ").lower()
if color == "red":
    print("Stop")
elif color == "yellow":
    print("Slow down")
elif color == "green":
    print("Go")
else:
    print("Invalid color")
```

## 9. Greatest Among Three Numbers

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))
if a >= b and a >= c:
    print("Greatest =", a)
elif b >= a and b >= c:
    print("Greatest =", b)
else:
    print("Greatest =", c)
```

## 10. Smallest Among Three Numbers

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))
if a <= b and a <= c:
    print("Smallest =", a)
elif b <= a and b <= c:
    print("Smallest =", b)
else:
    print("Smallest =", c)
```
