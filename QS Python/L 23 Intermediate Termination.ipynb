## **Intermediate Termination**

* These statements are used to **alter the normal flow of execution** within a loop.
* Python provides three keywords for this purpose:

  1. `break`
  2. `continue`
  3. `pass`

### **`break`**

* The `break` statement is used **inside a loop**.
* It **immediately terminates the loop** and transfers control to the statement after the loop.
* It can be used with both `while` and `for` loops.
* It is used with a **condition**

**Syntax:**

```python
# while loop

while condition:
    if condition:
        break

# for loop

for var in iterable:
    if condition:
        break
```
> **NOTE:** `break` and `continue` will not execute any statement writen imidiately after them.

````python
# Example 

l1=[1,2,'hii',3,4,5,True]
for i in l1:
    if type(i)==str:
        print(i)
        break
print(i)
````

````python
for i in range(1,6):
    if i==3:
        break
    print(i)
````

### **`continue`**

* The `continue` statement is used to **skip the current iteration** of a loop.
* After executing `continue`, control moves to the **next iteration** of the loop.
* It is used **inside a loop** and can be used with both `for` and `while` loops.
* In a `while` loop, the **updation should be performed before `continue`** to avoid an infinite loop.

### **Syntax**

**Using `for` loop:**

```python
for var in iterable:
    if condition:
        continue
```

**Using `while` loop:**

```python
while condition:
    if condition:
        updation
        continue
    updation
```

> **Note:** if we don't use updation inside a condition before `continue`  it will get in an infinite loop.


````python
#Example

for i in range(1, 6):
    if i == 3:
        continue
    print(i)
````

### **`pass`**

* `pass` is a **keyword** used as a placeholder when a statement block is required but no action needs to be performed.
* It allows an **empty block to be syntactically valid**.
* It does **not terminate or skip** the execution of a loop.
* It can be used inside **conditional statements, loops, functions, and classes**.

### **Syntax**

```python
if condition:
    pass
```


````python
# Example

for i in range(1, 6):
    if i == 3:
        pass
    print(i)
````
