```python
# 9. Wap to print factorial a number.
n=int(input("Enter n"))
fact=1
while n>=1:
    fact*=n
    n-=1
print(fact)
```

```python
# 12.	Wap to reverse a number without using slicing
n=int(input("number"))
while n>=1:
    if str(n)==str(n)[::-1] and len(str(n))>=2:
        print(n)
    n-=1
```
