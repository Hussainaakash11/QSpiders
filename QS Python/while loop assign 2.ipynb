## 8. Print n natural palindrome numbers

```python
n=int(input("Enter count: "))
num=1
count=0
while count<n:
    temp=num 
    rev=0
    while temp>0:
        rev=rev*10+temp%10
        temp//=10
    if num==rev:
        print(num)
        count+=1
    num+=1
```

## 20. Check Armstrong number

```python
n=int(input("Enter number: "))
original=n
power=len(str(n))
total=0
while n>0:
    total+=(n%10)**power
    n//=10
if total==original: 
    print("Armstrong number")
else: 
    print("Not an Armstrong number")
```

## 21. Check Disarium number

```python
n=int(input("Enter number: "))
original=n
pos=len(str(n))
total=0
while n>0:
    total+=(n%10)**pos
    pos-=1
    n//=10
if total==original: 
    print("Disarium number")
else: 
    print("Not a Disarium number")
```

## 23. Check Spy number

```python
n=int(input("Enter number: "))
total=0
product=1
while n>0:
    digit=n%10
    total+=digit
    product*=digit
    n//=10
if total==product: 
    print("Spy number")
else: 
    print("Not a Spy number")
```

## 24. Check Perfect number

## 25. Check Strong number

```python
n=int(input("Enter number: ")); original=n; total=0
while n>0:
    digit=n%10; fact=1; i=1
    while i<=digit:
        fact*=i; i+=1
    total+=fact; n//=10
if total==original: print("Strong number")
else: print("Not a Strong number")
```

## 26. Check Xylem or Phloem

```python
n=int(input("Enter number: ")); last=n%10; temp=n//10; middle=0
while temp>9:
    middle+=temp%10; temp//=10
first=temp
if first+last==middle: print("Xylem number")
else: print("Phloem number")
```

## Nested Loop 1. Extract prime numbers from a list

```python
numbers=[2,4,5,7,10,11,13,15,17]; result=[]; i=0
while i<len(numbers):
    n=numbers[i]; d=1; count=0
    while d<=n:
        if n%d==0: count+=1
        d+=1
    if count==2: result.append(n)
    i+=1
print(result)
```

## Nested Loop 2. Extract Armstrong numbers from a list

```python
numbers=[153,10,370,25,407,123]; result=[]; i=0
while i<len(numbers):
    n=numbers[i]; original=n; power=len(str(n)); total=0
    while n>0:
        total+=(n%10)**power; n//=10
    if total==original: result.append(original)
    i+=1
print(result)
```

## Nested Loop 3. Extract Perfect numbers from a tuple

```python
numbers=(6,10,28,12,496,20); result=[]; i=0
while i<len(numbers):
    n=numbers[i]; d=1; total=0
    while d<n:
        if n%d==0: total+=d
        d+=1
    if total==n: result.append(n)
    i+=1
print(result)
```

## Nested Loop 4. Extract Xylem numbers from a list

```python
numbers=[1234,123,246,121,12345]; result=[]; i=0
while i<len(numbers):
    n=numbers[i]; last=n%10; temp=n//10; middle=0
    while temp>9:
        middle+=temp%10; temp//=10
    if temp+last==middle: result.append(n)
    i+=1
print(result)
```
