## **Input & Output Operations**

Python provides two commonly used built-in functions for performing **input and output operations**:

1. `input()`
2. `print()`

### **`print()` Function**

* The `print()` function is used to **display data or output** on the screen.
* It can display **one or multiple values**.

**Syntax:**

```python
print(val1, val2, ..., valn)
```

**Example:**

```python
name = "Aakash"
age = 22

print(name, age)
```


````python
a=10
print(a)
print("Hii")
````

## **`input()` Function**

* The `input()` function is used to **take input from the user**.
* By default, `input()` returns the entered value as a **string (`str`)**.
* **Type casting** can be used when input is required in another data type.

### **1. Manual Type Casting**

The required data type is explicitly specified using a type-casting function.

**Syntax:**

```python
var = data_type(input("Instruction: "))
```

**Example:**

```python
age = int(input("Enter your age: "))
```

> **Note:** Manual type casting can be used to convert input into different data types, provided the entered value is compatible with the target type.

### **2. `eval()`**

* `eval()` evaluates the entered input as a **Python expression** and returns the resulting value.
* It can accept expressions that produce both **single-value and collection data types**.

**Syntax:**

```python
var = eval(input("Message: "))
```

**Example:**

```python
data = eval(input("Enter a value: "))
```

> **Note:** `eval()` should be used carefully with untrusted input because it can execute arbitrary Python expressions.


````python
# Example

a=int(input("Enter no:" ))
print(a)
print(type(a))
````

````python
a=eval(input("Enter the Data:"))
print(a)
print(type(a))
````

### **`end` and `sep` Arguments in `print()`**

The `print()` function provides two commonly used arguments: **`sep`** and **`end`**.

### **1. `sep` Argument**

* `sep` stands for **separator**.
* It specifies what should be placed **between multiple values**.
* The default value of `sep` is a **space (`" "`)**.

**Syntax:**

```python
print(val1, val2, ..., valn, sep="separator")
```

**Example:**

```python
print("Aakash", "Hussain", sep="-")
# Aakash-Hussain
```

### **2. `end` Argument**

* `end` specifies what should be printed **at the end of the output**.
* The default value of `end` is a **newline (`"\n"`)**.

**Syntax:**

```python
print(val1, val2, ..., valn, end="ending")
```

**Example:**

```python
print("Hello", end=" ")
print("World")
# Hello World
```

> **Note:** `sep` works **between values**, while `end` works **after all values** have been printed.


````python
print(10,20,3,40,50,sep='-')
````

````python
print("Hii",end=' ')
print("Hello",end=' ')
print("how are you",end=' ')
````
