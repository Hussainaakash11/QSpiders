## **Regular Expression**
- It is a sequence of characters that defines a search pattern used to search, match, validate, or replace patterns within text (string data).
- We use the built-in `re` module to work with regular expressions in Python.
- `search()`: Searches the entire string and returns a match object for the first occurrence of the pattern (or `None` if no match is found).
- `match()`: Checks for a match only at the beginning of the string and returns a match object (or `None` if no match is found).
- `sub()`: Replaces occurrences of the matched pattern in the string with a new replacement string.
- `findall()`: Returns a list of all non-overlapping matches of the pattern found in the string.
