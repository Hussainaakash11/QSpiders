## **Regular Expression**
- It is a sequence of characters that defines a search pattern used to search, match, validate, or replace patterns within text (string data).
- We use the built-in `re` module to work with regular expressions in Python.
- `search()`: Searches the entire string and returns a match object for the first occurrence of the pattern (or `None` if no match is found).
- `match()`: Checks for a match only at the beginning of the string and returns a match object (or `None` if no match is found).
- `sub()`: Replaces occurrences of the matched pattern in the string with a new replacement string.
- `findall()`: Returns a list of all non-overlapping matches of the pattern found in the string.
- Syntax:
```python
import re 
re.findall("exp/pattern", string_name)
```

````python
import re 
s = "this is a string"
out = re.search(r"string", s)
output = re.match(r"string", s)
print(out)
print(output)
````

````python
s= "Hello"
out = re.sub("l", "k", s)
print(out)
````

````python
s = "hello"
out = re.findall(r"l", s)
print(out)
````

### **Special Characters (Metacharacters)**
- `.`: Matches any single character (except a newline `\n`).
- `^`: Matches the start of the string (or the start of each line in multiline mode).
- `$`: Matches the end of the string (or just before a newline at the end of the string).
- `*`: Matches 0 or more repetitions of the preceding character or group.
- `+`: Matches 1 or more repetitions of the preceding character or group.
- `?`: Matches 0 or 1 repetition of the preceding character only (also makes quantifiers non-greedy).
- `{m,n}`: Matches between `m` and `n` repetitions of the preceding character or group.
- `[]`: Matches any single character inside the brackets (character set/range, e.g., `[a-z]`, `[0-9]`).
- `[^]`: Matches any single character **not** present inside the brackets (negated set).
- `|`: Acts as a logical OR operator, matching either the pattern before or after it.
- `()`: Groups patterns together and captures the matched substring.
- `\`: Escapes special characters or signals a special sequence (e.g., `\d`, `\w`, `\s`).

### **Special Sequences**
- `\d`: Matches any digit (`[0-9]`).
- `\D`: Matches any non-digit character (`[^0-9]`).
- `\w`: Matches any alphanumeric character or underscore (`[a-zA-Z0-9_]`).
- `\W`: Matches any non-alphanumeric character (`[^a-zA-Z0-9_]`).
- `\s`: Matches any whitespace character (spaces, tabs, newlines).
- `\S`: Matches any non-whitespace character.
- `\b`: Matches a word boundary (the position between a word and a non-word character).
- `\B`: Matches a position that is not a word boundary.

````python
s= "cat cut cbt coot coat"
out = re.findall("c.t", s)
print(out)
````

````python
s = 'there hello'
out= re.findall('^hello',s)
print(out)
````

````python
s = 'hello There'
out= re.findall('hello$',s)
print(out)
````

````python
s = 'hello loo lo'
out= re.findall('o*',s)
print(out)
````

````python
s = 'hello loo lo'
out= re.findall('lo+',s)
print(out)
````

````python
s = 'hello loo lo'
out= re.findall('lo?',s)
print(out)
````

````python
s = 'color colour'
out= re.findall('colou?r',s)
print(out)
````

````python
s = 'Insomaniac'
out= re.findall('[aeiou]',s)
print(out)
````

````python
s = 'Insomaniac'
out= re.findall('[^aeiou]',s)
print(out)
````

````python
s = 'Insomaniac@123'
out= re.findall('[A-z]',s)
print(out)
````
