## **Multiple Value / Collection Data Types**

* Multiple-value data types can store **multiple values** in a single variable.
* They are generally **iterable**, allowing values to be accessed one by one.

### **1. String (`str`)**

* A string is a **sequence of textual data**.
* It can contain alphabets, alphanumeric characters, numbers, and special characters.

**Syntax:**

```python
var = 'val1, val2, val3'     # Single quotes
var = "val1, val2, val3"     # Double quotes
var = """val1, val2, val3""" # Triple quotes; supports multiple lines
```

### **Characteristics of String (`str`)**

* **Ordered** data type.
* Allows **duplicate values**.
* **Immutable** in nature.
* Supports **indexing and slicing**.
* An empty string is represented as `""`.

### **Indexing**

* Indexing is the process of accessing a **single character/value** from a collection using its index position.
* **Positive indexing:** Starts from `0` and goes up to `len() - 1`, usually from **left to right (L2R)**.
* **Negative indexing:** Starts from `-1` and goes up to `-len()`, usually from **right to left (R2L)**.

**Syntax:**

```python
var[index]
```

**Example:**

```python
name = "Python"

name[0]   # 'P'
name[-1]  # 'n'
```


````python
# Examples
s='Aakash'
x= "It's my Big Day"
z='''We're in Noida
inside qspiders'''
print(s)
print(x)
print(z)
````

### **Length `len()` Function**

`len(var)` : Returns the **total number of elements** in a collection or multiple-value data type.

### **Example:**

```python
x = "Aakash"

print(len(x))  # 6
```

````python
x = "Aakash"
print(len(x))
````

### **Slicing**

* Slicing is the process of **fetching multiple values/elements** from a collection using a range of indices.

### **Syntax**

```python
var[i:j:k]
```

Where:

* `i` = **Start index**
* `j` = **End index**
* `k` = **Step/Jump value**

### **Basics**

* `i` (start) is **included**.
* `j` (end) is **excluded**.
* `k` determines the **step/jump** between elements.

### **Default Values**

* `i` = `0` (positive step) / `-1` (negative step)
* `j` = `len(collection)` (positive step) / `-len(collection)-1` (negative step)
* `k` = `1`

### **Important**

* **Copying:** `var[::]`
* **Reversing:** `var[::-1]`

### **Example**

```python
name = "Python"

print(name[:4])  # Pyth
```

> **Note:** The end index is always excluded.
