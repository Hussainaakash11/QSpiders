## **Recursion**

- **Recursion** is a technique in which a function **calls itself repeatedly** until a **base condition** is satisfied.
- It can be used as an alternative to **looping** for solving certain problems.
- Recursion is commonly used in problems involving **trees, graphs, backtracking, and divide-and-conquer algorithms**.
- Each recursive call is stored in the **call stack**, following the **LIFO (Last In, First Out)** mechanism.
- Python has a default **recursion limit of approximately 1000 calls** to prevent excessive stack usage.
- The recursion limit can be modified using the `sys.setrecursionlimit()` function when required.

### **Basic Structure**

```python
def func(n):
    if base_condition:
        return

    func(n - 1)
```

**Note:** A recursive function must have a **base condition**; otherwise, it may result in infinite recursion.

````python
import sys
print(sys.getrecursionlimit())
sys.setrecursionlimit(3000)
print(sys.getrecursionlimit())
````
