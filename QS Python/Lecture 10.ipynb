## **Set**

* A **set** is a collection of **unique values**.
* It is mainly used when duplicate values need to be removed or when performing set operations.

**Syntax:**

```python
var = {val1, val2, val3, ..., valn}
```

### **Characteristics**

1. **Unordered** data type.
2. Does **not allow duplicate values**.
3. Does **not support indexing or slicing**.
4. It is **mutable**, but its elements must be **immutable/hashable**.
5. Empty set: `set()`

> **Note:** `{}` represents an **empty dictionary**, not an empty set. Use `set()` to create an empty set.

````python
#Empty Set
set1=set()
print(len(set1))
print(type(set1))
````

````python
set2={6,1,5,88,5,6,9,5,8,5,8,5,55,44,77,22,55,66,88,77,11}
set2
````

### **Some Built-in Set Methods**

1. **`add()`**

   * Adds a **single value** to the set.
   * Syntax: `var.add(value)`

2. **`update()`**

   * Adds **multiple values** to the set.
   * Syntax: `var.update({val1, val2, ..., valn})`

3. **`remove()`**

   * Removes a specified value from the set.
   * Raises a **`KeyError`** if the value does not exist.
   * Syntax: `var.remove(value)`

4. **`pop()`**

   * Removes and returns an **arbitrary element** from the set.
   * Syntax: `var.pop()`

5. **`clear()`**

   * Removes **all elements** from the set.
   * Syntax: `var.clear()`


````python
set2={1,2,3,6,5,4,2.2,3.3,55,False,"hii",(2,6,5,8)}
# Adding an element to the set
set2.add(100)
print(set2)

# Removing an element from the set
set2.remove(4)
print(set2)

# Updating the set with multiple elements
set2.update([1000,2000,3000])
print(set2)

# Popping an arbitrary element from the set
set2.pop()
print(set2)

# Clearing all elements from the set
set2.clear()
print(set2)
````

## **Dictionary**

* A **dictionary** is a collection of **key-value pairs**, where each key and its corresponding value together form an **item**.
* Each **key must be unique** within a dictionary.
* Dictionaries are useful for storing and accessing data using meaningful **keys**.

**Syntax:**

```python
var = {
    key1: val1,
    key2: val2,
    ...
    keyN: valN
}
```

### **Characteristics**

1. Stores data in **key-value pairs**.
2. **Keys must be unique**, but values can be duplicated.
3. **Ordered** data type (preserves insertion order).
4. **Mutable** in nature.
5. Values can be accessed using their **keys**, not index positions.
6. Empty dictionary: `{}`

````python
#Example of dictionary
dict1={"name":"Hussain",
       "age":22,
       "city":"Bangalore"}
print(dict1)

#Length of the dictionary
print(len(dict1))
````
