## **Built-in Methods of String:**

String methods are **built-in methods used to perform various operations on string values**.

### **1. `upper()`**

Converts all alphabetic characters of a string to uppercase.

**Syntax:** `var.upper()`

### **2. `lower()`**

Converts all alphabetic characters of a string to lowercase.

**Syntax:** `var.lower()`

### **3. `swapcase()`**

Converts uppercase characters to lowercase and lowercase characters to uppercase.

**Syntax:** `var.swapcase()`

### **4. `capitalize()`**

Converts the first character to uppercase and the remaining alphabetic characters to lowercase.

**Syntax:** `var.capitalize()`

### **5. `title()`**

Converts the first character of each word to uppercase and the remaining alphabetic characters to lowercase.

**Syntax:** `var.title()`

### **6. `isupper()`**

Checks whether all alphabetic characters in the string are uppercase.

**Syntax:** `var.isupper()`

### **7. `islower()`**

Checks whether all alphabetic characters in the string are lowercase.

**Syntax:** `var.islower()`

### **8. `istitle()`**

Checks whether the string follows title-case formatting.

**Syntax:** `var.istitle()`

### **9. `isdigit()`**

Checks whether all characters in the string are digits.

**Syntax:** `var.isdigit()`

### **10. `isalpha()`**

Checks whether all characters in the string are alphabetic.

**Syntax:** `var.isalpha()`

### **11. `isalnum()`**

Checks whether all characters in the string are alphanumeric (letters or digits).

**Syntax:** `var.isalnum()`

### **12. `isspace()`**

Checks whether all characters in the string are whitespace characters.

**Syntax:** `var.isspace()`

### **13. `startswith()`**

Checks whether the string starts with the specified substring.

**Syntax:** `var.startswith(value)`

### **14. `endswith()`**

Checks whether the string ends with the specified substring.

**Syntax:** `var.endswith(value)`


```python
s="KashMir is beautiful"
print("Title:", s.title())
print("Upper:", s.upper())
print("Lower:", s.lower())
print("Capitalize:", s.capitalize())
print("Swapcase:", s.swapcase())
print("Count of 'i':", s.count('i'))
print("Is upper:", s.isupper())
print("Is lower:", s.islower())
print("Is title:", s.istitle())
print("Is alpha:", s.isalpha())
print("Is alphanumeric:", s.isalnum())
print("Is digit:", s.isdigit())
print("Is space:", s.isspace())
print("Starts with 'Kash':", s.startswith('Kash'))
print("Ends with 'beautiful':", s.endswith('beautiful'))
```

### **15. `replace()`**

Replaces all occurrences of an old substring with a new substring.
**Syntax:** `var.replace(old_str, new_str)`

### **16. `count()`**

Returns the number of occurrences of a substring in the string.
**Syntax:** `var.count(substring)`

### **17. `index()`**

Returns the first index of the specified substring and raises `ValueError` if it is not found.
**Syntax:** `var.index(substring)`

### **18. `rindex()`**

Returns the last index of the specified substring and raises `ValueError` if it is not found.
**Syntax:** `var.rindex(substring)`

### **19. `find()`**

Returns the first index of the specified substring and returns `-1` if it is not found.
**Syntax:** `var.find(substring)`

#### **`index()` vs `find()`**

| Method    | If Found      | If Not Found |
| --------- | ------------- | ------------ |
| `index()` | Returns index | `ValueError` |
| `find()`  | Returns index | `-1`         |

### **20. `rfind()`**

Returns the last index of the specified substring and returns `-1` if it is not found.
**Syntax:** `var.rfind(substring)`

### **21. `strip()`**

Removes leading and trailing whitespace or specified characters from a string.
**Syntax:** `var.strip(chars)`

### **22. `lstrip()`**

Removes leading whitespace or specified characters from the left side of a string.
**Syntax:** `var.lstrip(chars)`

### **23. `rstrip()`**

Removes trailing whitespace or specified characters from the right side of a string.
**Syntax:** `var.rstrip(chars)`

### **24. `split()`**

Splits a string into multiple parts and returns the result as a list.
**Syntax:** `var.split(separator)`

> If no separator is specified, the string is split based on whitespace.

### **25. `join()`**

Joins elements of an iterable into a single string using the specified separator.
**Syntax:** `separator.join(iterable)`

> The separator is written before `.join()`.
