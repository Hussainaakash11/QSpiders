## **Polymorphism**

- **Polymorphism** is the ability to use the **same operator or method** to perform different operations depending on the context.
- Example: The `+` operator performs **addition** with numbers and **concatenation** with strings.

### **Types of Polymorphism**

There are four common ways to achieve polymorphism in Python:

1. **Method Overloading**
2. **Method Overriding**
3. **Duck Typing**
4. **Operator Overloading**

## **1. Method Overloading**

- **Method overloading** is the concept of using the **same method name with different numbers or types of arguments**.
- Python does **not support true method overloading** because defining multiple methods with the same name causes the **latest definition to replace the previous one**.
- However, similar behavior can be achieved using flexible arguments.

### **Ways to Achieve Method Overloading in Python**

1. Using **Default Arguments**
2. Using **`*args` and `**kwargs`**

````python
# Using **Default Arguments**
class Demo:
    def add(self,a,b,c=0,d=0):
        print(a+b+c+d)

    # def add(self,a,b,c):
    #     print(a+b+c)

    # def add(self,a,b,c,d):
    #     print(a+b+c+d)

obj=Demo()
obj.add(2,3)
obj.add(2,63,2)
obj.add(2,63,2,5)
````

````python
# Using **`*args` and `**kwargs`**

class Values:

    def add(self,*args):
        if len(args)==2:
            print("product is : ", args[0]*args[1])
        elif len(args)>2:
            print("Sum is: ",sum(args))
        elif len(args)==1:
            print("Sq is:", args[0]**2)

obj=Values()

obj.add(2)
obj.add(2,6)
obj.add(2,6,5)
````

## **2. Method Overriding**
- **Method overriding** is the process of redefining a **parent class method** in the child class according to the requirements of the child class.
- **Inheritance is required** for method overriding.
- The **method name and parameters** should be compatible between the parent and child methods.
- **Method Signature** = Method name + Parameters.

### **Example**

```python
class Parent:
    def show(self):
        print("Parent method")

class Child(Parent):
    def show(self):
        print("Child method")

obj = Child()
obj.show()
```

**Output:**
```text
Child method
```

````python
class Parent:
    def marry(self):
        print("Marry to ABC")

class Child(Parent):
    def marry(self):
        print("Marry to XYZ")

obj=Child()
obj.marry()
````

````python
class SBI:
    def greet(self):
        print("Hello: ")
        print("WelCome to SBI")

    def deposit(self):
        print("Proceed to Deposit")

    def withdraw(self):
        print("proceed to withdraw")

class ICICI(SBI):
    def greet(self):
        print("Hello: ")
        print("WelCome to ICICI")

c1=SBI()
c2=ICICI()

c1.greet()
c2.greet()
        
````

## **3. Duck Typing**

- **Duck Typing** is based on the principle: *“If it walks like a duck and quacks like a duck, then it is a duck.”*
- It means **the behavior of an object matters more than its class or type**.
- Multiple unrelated classes can have the **same method**, and Python can use the method based on the object's behavior.
- No inheritance or relationship between the classes is required.

### **Example**

```python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

def make_sound(animal):
    animal.sound()

make_sound(Dog())
make_sound(Cat())
```

Here, `make_sound()` works with both objects because both provide the required `sound()` method.

````python
class Animal:
    def speak(self):
        print("It makes a sound")

class Human:
    def speak(self):
        print("It Speakes")

class Cat:
    def speak(self):
        print("Meow")

class Dog:
    def speak(self):
        print("It Barks")


obj1 = Animal()
obj2 = Human()
obj3 =Dog()
obj4 = Cat()

def say_something(obj):
    obj.speak()

say_something(obj2)
say_something(obj4)
````

## **4. Operator Overloading**

- **Operator overloading** is the process of giving a **different meaning to the same operator** depending on the objects or operands involved.
- In Python, operator overloading is achieved using **special methods (magic/dunder methods)** such as `__add__()`, `__sub__()`, `__mul__()`, etc.
- Example: The `+` operator performs addition for numbers and concatenation for strings.

### **Syntax**

```python
class ClassName:
    def __add__(self, other):
        # Define the behavior of +
        return result
```

### **Common Operator Methods**

| Operator | Special Method |
|---|---|
| `+` | `__add__()` |
| `-` | `__sub__()` |
| `*` | `__mul__()` |
| `/` | `__truediv__()` |
| `//` | `__floordiv__()` |
| `%` | `__mod__()` |
| `**` | `__pow__()` |
| `==` | `__eq__()` |
| `<` | `__lt__()` |
| `>` | `__gt__()` |

````python
class Arthematic:
    def __init__(self,val):
        self.a = val

    def __add__(self, other):
        return self.a + other.a
    
ob1=Arthematic(10)
ob2=Arthematic(20)

print(ob1+ob2)
````

### Difference Between the Four Types of Polymorphism

| Type | Simple Meaning | Inheritance Required? | Main Idea |
|---|---|---|---|
| **Method Overloading** | Same method, different number/type of arguments | No | One method handles different arguments |
| **Method Overriding** | Child class changes the parent class method | **Yes** | Same method, different implementation |
| **Duck Typing** | Behavior matters, not the class | No | If an object has the required method, it can be used |
| **Operator Overloading** | Same operator performs different operations | No | Operators behave differently for different objects |

**Easy way to remember:**

- **Overloading** → Same method, **different arguments**
- **Overriding** → Same method, **different implementation**
- **Duck Typing** → **Behavior** matters
- **Operator Overloading** → Same operator, **different meaning**
