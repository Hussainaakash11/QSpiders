## **Package Architecture**

### **Package vs. Module**

| Package | Module |
| :--- | :--- |
| It is a directory (folder) that contains `n` number of modules (and sub-packages) inside it. | It is a Python file (`.py`) that contains `n` number of instructions, functions, classes, and variables inside it. |
| Third-party packages must be installed before importing; built-in packages can be imported directly. | To use a module, we have to import it into our program. |
| Examples: `NumPy`, `Pandas`, `Django`, etc. | Examples: `keyword`, `math`, `random`, `copy`, etc. |
| Installed via package manager: `pip install package_name` | Imported via statement: `import module_name` |

### **Ways to Import Module:**
- In three ways we can import module:
1. By using import keyword.
2. By using 'from' and 'import' keyword.
3. By using '*'

#### **1. By using import keyword:**
- In this we can import all the functions/methods present in the module.
- It consumes more space.
- Syntax:
```python
import module_name
module_name.function1()
module_name.function2()
```

````python
import math
print(math.pi)
print(math.e)
print(math.sqrt(16))
print(math.factorial(5))
print(math.pow(2,5))
````

#### **2. By using 'from' and 'import' keyword:**
- In this we can import only reqiured function/method present in the module.
- It consumes less amount of space.
- Syntax:
```python
from module_name import function1, function2
function1()
function2()
```

````python
from math import factorial as fact, pi, pow
print(fact(5))
print(pow(5,2))
print(pi)
radius=3
area_of_circle = pi*radius**2
print((area_of_circle))
````

#### **3. By using '*'**
- In this we can import all the functions/methods present in the module.
- It consumes more space.
- Syntax:
```python
from module_name import *
function1()
function2()
```

````python
from math import *
print(pi)
print(e)
print(sqrt(16))
print(factorial(5))
print(pow(2,5))
````
