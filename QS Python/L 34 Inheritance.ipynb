## **Pillars of OOPs**

- The **four pillars of OOP** are fundamental concepts that provide structure and powerful features to object-oriented systems.
- They help in making programs more **modular, reusable, maintainable, and secure**.
- There are four pillars of OOP:

1. **Inheritance**
2. **Polymorphism**
3. **Encapsulation**
4. **Abstraction**

### **1. Inheritance**

- **Inheritance** is a mechanism that allows a **child class** to acquire the properties and methods of a **parent class**.
- It promotes **code reusability** and supports hierarchical classification.
- The class whose properties and methods are inherited is called the **Parent/Base/Super class**.
- The class that inherits them is called the **Child/Derived/Sub class**.
- Inheritance follows a **parent-to-child** relationship.

### **Types of Inheritance**

There are **five types of inheritance**:

1. **Single Inheritance**
2. **Multilevel Inheritance**
3. **Multiple Inheritance**
4. **Hierarchical Inheritance**
5. **Hybrid Inheritance**

#### **1. Single Inheritance**

- In **Single Inheritance**, a single **child class** inherits properties and methods from a single **parent class**.
- It represents a **one-to-one inheritance relationship**.

### **Syntax**

```python
class ParentClass:
    # Statement Block
    pass

class ChildClass(ParentClass):
    # Statement Block
    pass
```

Here, `ChildClass` inherits the properties and methods of `ParentClass`.

````python
class Demo():
    a=10
    b=20

    def show(self):
        print("This is demo method")

class Myclass(Demo):
    pass

ob1=Myclass()
print(ob1.a)
print(ob1.b)
ob1.show()
````

#### **2. Multilevel Inheritance**

- In **Multilevel Inheritance**, properties and methods are inherited through **multiple levels of inheritance**.
- A class inherits from another class, and a further class inherits from the derived class.
- It forms a **chain of inheritance**.

### **Example**

```python id="n8j4st"
class GrandParent:
    pass

class Parent(GrandParent):
    pass

class Child(Parent):
    pass
```

Here, `Child` indirectly inherits the properties and methods of `GrandParent` through `Parent`.

````python
class Ten:
    sub1="math"
    sub2="eng"
class Twelvth(Ten):
    sub3="his"
    sub4="geo"
class Bachlor(Twelvth):
    sub5="urdu"
    sub6="kmri"
class Master(Bachlor):
    sub7="CS"
    sub8="Bio"

ob1=Master()
print(ob1.sub1)
print(ob1.sub3)
````

#### **3. Multiple Inheritance**

- In **Multiple Inheritance**, a single **child class** inherits properties and methods from **multiple parent classes**.
- The child class can access the members inherited from all its parent classes.

#### **Syntax**

```python
class Parent1:
    pass

class Parent2:
    pass

class Child(Parent1, Parent2):
    pass
```

- Here, `Child` inherits from both `Parent1` and `Parent2`.

````python
class Add:
    @staticmethod
    def add(a,b):
        return a+b

class Sub:
    @staticmethod
    def sub(a,b):
        return a-b

class Prod:
    @staticmethod
    def mul(a,b):
        return a*b

class Calci(Add,Sub,Prod):
    @staticmethod
    def div(a,b):
        return a/b

obj=Calci()

print(obj.add(10,6))
print(obj.sub(6,3))
print(obj.mul(6,6))
print(obj.div(9,6))
````

##### **MRO (Method Resolution Order)**

- **MRO** defines the order in which Python searches for an **attribute or method** in a class hierarchy, especially in **multiple inheritance**.
- Python first searches in the **child class**, then follows the defined inheritance hierarchy.
- In multiple inheritance, Python generally follows the **left-to-right order** of the parent classes while maintaining a consistent hierarchy.
- MRO can be viewed using the `mro()` method or `__mro__` attribute.

### **Example**

```python id="m0l8cy"
class A:
    pass

class B:
    pass

class C(A, B):
    pass

print(C.mro())
```

Here, Python searches `C` first, followed by `A`, then `B`, according to the MRO.

````python
class Add:
    @staticmethod
    def add(a):
        return a+2

class Sub:
    @staticmethod
    def add(a):
        return a+3

class Prod:
    @staticmethod
    def add(a):
        return a+4

class Calci(Add,Sub,Prod):
    @staticmethod
    def addd(a):
        return a+5

obj=Calci()

# print(Calci.mro())

print(obj.add(10))
````

### **4. Hierarchical Inheritance**

- In **hierarchical inheritance**, a single parent class is inherited by **multiple child classes**.
- It represents a **one-to-many** inheritance relationship.

### **Syntax**

```python
class Parent:
    # Statement Block
    pass

class Child1(Parent):
    # Statement Block
    pass

class Child2(Parent):
    # Statement Block
    pass
```

Here, both `Child1` and `Child2` inherit the properties and methods of the `Parent` class.

````python
class Animal:
    def speak(self):
        print("It makes sound")

class Human(Animal):
    pass

class Dog(Animal):
    pass 

class Cat(Animal):
    pass

obj=Cat()

obj.speak()
````

### **5. Hybrid Inheritance**

- **Hybrid inheritance** is a combination of **two or more types of inheritance**.
- It combines different inheritance structures to form a more complex hierarchy.

### **Syntax**

```python
class A:
    pass

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass
```

Here, the combination of **hierarchical and multiple inheritance** forms hybrid inheritance.
