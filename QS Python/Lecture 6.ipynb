## **Identifier**

* An **identifier** is the name used to identify elements in Python, such as **variables, functions, methods, classes, and objects**.
* Python follows certain rules for naming identifiers.

### **Rules of Identifiers**

1. **Keywords** cannot be used as identifiers.
2. An identifier **cannot begin with a number**.
3. **Spaces** are not allowed in identifiers.
4. **Special characters** are not allowed, except the underscore (`_`).
5. Identifiers are **case-sensitive** (`name` and `Name` are different).
6. Avoid using lowercase **`l`** and uppercase **`O`** when possible, as they can be confused with **`1`** and **`0`**.

## **Data Types**

Data types specify the **type of value** a variable can store and determine the operations that can be performed on it.

### **1. Single Value / Individual Data Types**

* These generally represent a **single value** and are non-iterable.
* **int** – Integer values
* **float** – Decimal values
* **complex** – Complex numbers
* **bool** – Boolean values (`True` or `False`)

### **2. Multiple Value / Collection Data Types**

* These store **multiple values** and are generally iterable.
* **str** – Sequence of characters
* **list** – Ordered, mutable collection
* **tuple** – Ordered, immutable collection
* **set** – Unordered collection of unique values
* **dict** – Collection of key-value pairs


### **Single Value / Individual Data Types**

* These data types generally store **a single value in a memory location**.

**1. Integer (`int`):**

* Represents **positive and negative whole numbers**, including zero.
* Examples: `1`, `-2`, `45`, `556`, `0`

**2. Float (`float`):**

* Represents **real numbers with decimal values**.
* Examples: `1.23`, `3.69`, `55.25`

**3. Complex Number (`complex`):**

* Represents a combination of **real and imaginary parts**.
* General form: `a + bj`
* Examples: `3 + 2j`, `45 + 69j`
* `j` is used to represent the **imaginary part**.

**4. Boolean (`bool`):**

* Represents one of two values: **`True` or `False`**.
* Internally, `True` corresponds to `1` and `False` corresponds to `0`.
* `True` and `False` are **Python keywords** used in logical and decision-making operations.

### **`type()` Function**

`type(val/var)` : Returns the **data type (class)** of the given value or variable.

**Example:**

```python
x = 10
print(type(x))   # <class 'int'>
```

````python
a=90
print(type(a))
b=90.87
print(type(b))
c=22
print(type(c))
````
