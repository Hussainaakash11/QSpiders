## **OOPs:** *Object-Oriented Programming*

- OOP stands for **Object-Oriented Programming**.
- It is a programming paradigm where programs are designed using **objects and classes**.
- OOP helps in solving **real-world problems** by representing entities as objects.
- While developing a project, **class-based programming** helps organize data and functionality effectively.

### **Class**

- A **class** is a blueprint or template for creating objects.
- It defines the **attributes (data)** and **behaviours (methods)** of objects.
- It provides a structure through which data and functionality can be defined and accessed.

### **Object**

- An **object** is a real-world entity or representation created from a class.
- It is an **instance of a class**.
- An object contains **attributes (properties)** and **behaviours (methods)** defined by its class.
- Every object belongs to a class.
- A single class can have **multiple objects**.

### **Syntax of Class & Object**

```python
class ClassName:
    # Attributes
    # Behaviours

obj1 = ClassName()
obj2 = ClassName()
```

- The `class` keyword is used to define a class.
- `ClassName` is the name of the class.
- `obj1` and `obj2` are **objects (instances)** of the class.
- `ClassName()` is used to **create an object** of the class.

> **Attributes:** Properties / States / Members / Data / Variables

> **Behaviours:** Features / Functions / Methods

> **Note:** A class can be considered a **user-defined data type**, as it defines the structure and behaviour of objects.

### **Types of Data Types**

1. **Built-in Data Types** — predefined by Python.
   - Examples: `int`, `float`, `str`, `list`, `tuple`, `set`, `dict`

2. **User-Defined Data Types** — created by the programmer using classes.
   - Example: `class Student:`

### **Attributes / Properties**

- Data or variables associated with a class or object are known as **attributes/properties**.
- There are two types of properties:
  1. **Class Properties**
  2. **Object Properties**

#### **1. Class Properties**

- Class properties are the attributes that are **shared by all objects** of a class.
- They provide information that is **common to the entire class**.
- A variable defined directly inside a class, outside methods, is generally considered a **class property**.
- They can be accessed using the **class name or object name**.

**Example:**  
If `Bank` is a class, its common class properties can be:

- `name`
- `IFSC Code`
- `CEO`
- `Manager`
- `Interest Rate`

```python
class Bank:
    name = "SBI"
    interest_rate = 7.5
```

Here, `name` and `interest_rate` are **class properties** shared by all objects of the `Bank` class.

### **How to access class Properties:**
```python
className.Propertyname

obj.propertyname
```

### **How to modify class Properties:**
```python

```

#### **2. Object Properties**

- Object properties are the attributes whose **values can be different for each object**.
- They provide information specific to a particular **object/instance**.
- They are generally defined inside the **constructor `__init__()`** using `self`.
- Each object gets its own copy of these properties.

**Example:**  
If customers are objects of the `Bank` class, object properties can include:

- Customer's name
- Phone number
- Address
- Account number
- Aadhaar number
- Balance

```python id="u2gk8m"

        self.name = name
        self.account_no = account_no
        self.balance = balance
```

Here, `name`, `account_no`, and `balance` are **object properties**, and their values can be different for each object.

### **Constructor / `__init__()`**

- `__init__()` is a special method used to **initialize the object properties**.
- It is used to define and initialize **object-specific attributes**.
- It is **automatically invoked** when an object is created; we normally do not call it explicitly.
- `self` refers to the **current object** and is used to access or initialize its properties.
- The constructor can accept additional parameters to initialize object properties.

### **Syntax**

```python
class ClassName:
    def __init__(self, var1, var2, ..., varN):
        self.prop1 = var1
        self.prop2 = var2
        ...
        self.propN = varN

obj1 = ClassName(val1, val2, ..., valN)
obj2 = ClassName(val1, val2, ..., valN)
```

### **Example**

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Aakash", 21)
student2 = Student("Rahul", 20)
```

Here, `student1` and `student2` have their **own values** for `name` and `age`.

### **Functionalities / Methods / Behaviours**

- A **function defined inside a class** is called a **method**.
- A function is generally independent of a class, whereas a method is **associated with a class or object**.
- A function can be called directly, while a method is generally called using an **object or class**, depending on its type.
- There are two types of methods:

  1. **Built-in Methods**
     - Predefined methods provided by Python.
     - Examples: `append()`, `pop()`, `clear()`

  2. **User-Defined Methods**
     - Methods created by the programmer according to the requirements of the class.

### **2. User Defined Methods:**

- These are the methods defined by the user inside the user defined class according to the requirements.
- There are three types of User Defined Methods:
    1. Class Method.
    2. Object Method / Instance Method.
    3. Static Method.


#### **1. Class Method:**

- It is used to access and modify the class properties.
- An inbuilt decorator `classmethod` is used to define a class method.
- It takes an mandatory argument i.e, `cls` which carries the address of the class.
```python
class ClassName:
    @classmethod
    def method_name(cls):
        #Statement Block

ClassName.method_name(args)
```

### **3. Static Method**

- A **static method** is a utility/helper method used to store functionality that does not depend on a specific object or class state.
- It can be called using both the **class name and object**.
- The built-in `@staticmethod` decorator is used to define a static method.
- It does **not require `self` or `cls`** as a mandatory argument.
- It can be called without creating an object.

### **Syntax**

```python
class ClassName:
    @staticmethod
    def method_name():
        # Static method
        pass

obj1 = ClassName()

obj1.method_name()
ClassName.method_name()
```

````python
class Bank:
    # ---------------------------------------------------------
    # Class Properties
    # These properties are shared by all objects of the class.
    # ---------------------------------------------------------
    bank_name = "SBI"
    manager = "Maliya"
    location = "London"

    # ---------------------------------------------------------
    # Class Method: Access Class Properties
    # @classmethod allows us to work with class-level data.
    # 'cls' refers to the class itself.
    # ---------------------------------------------------------
    @classmethod
    def bank_info(cls):
        print("Bank Name     :", cls.bank_name)
        print("Bank Manager  :", cls.manager)
        print("Bank Location :", cls.location)

    # ---------------------------------------------------------
    # Class Method: Update Class Property
    # This method changes the location for the entire bank.
    # ---------------------------------------------------------
    @classmethod
    def change_bank_location(cls, new_location):
        cls.location = new_location
        print("Bank location changed successfully.")

    # ---------------------------------------------------------
    # Constructor
    # Initializes the properties of each customer object.
    # ---------------------------------------------------------
    def __init__(self, name, account_number, balance, nominee_name, pan_card):
        self.name = name
        self.account_number = account_number
        self.balance = balance
        self.nominee_name = nominee_name
        self.pan_card = pan_card

    # ---------------------------------------------------------
    # Instance Method: Display Customer Information
    # 'self' refers to the current customer object.
    # ---------------------------------------------------------
    def customer_info(self):
        print("\n----- Customer Information -----")
        print("Customer Name      :", self.name)
        print("Account Number     :", self.account_number)
        print("Account Balance    :", self.balance)
        print("Nominee Name       :", self.nominee_name)
        print("PAN Card           :", self.pan_card)

    # ---------------------------------------------------------
    # Instance Method: Change Account Number
    # Updates the account number of the current customer.
    # ---------------------------------------------------------
    def change_account_number(self, new_account_number):
        self.account_number = new_account_number
        print("Account number changed successfully.")

    # ---------------------------------------------------------
    # Instance Method: Deposit Money
    # Adds the entered amount to the customer's balance.
    # ---------------------------------------------------------
    def deposit(self):
        amount = self.get_amount()

        if amount <= 0:
            print("Deposit amount must be greater than zero.")
            return

        self.balance += amount
        print(f"₹{amount} deposited successfully.")
        print("Updated Balance :", self.balance)

    # ---------------------------------------------------------
    # Instance Method: Withdraw Money
    # Deducts the entered amount if sufficient balance exists.
    # ---------------------------------------------------------
    def withdraw(self):
        amount = self.get_amount()

        if amount <= 0:
            print("Withdrawal amount must be greater than zero.")
            return

        if amount <= self.balance:
            self.balance -= amount
            print(f"₹{amount} withdrawn successfully.")
            print("Remaining Balance :", self.balance)
        else:
            print("Insufficient balance.")
            print("Available Balance :", self.balance)

    # ---------------------------------------------------------
    # Static Method: Take Amount as Input
    # A static method does not depend on class or object data.
    # Therefore, it does not require 'self' or 'cls'.
    # ---------------------------------------------------------
    @staticmethod
    def get_amount():
        while True:
            try:
                amount = int(input("Enter the amount: "))

                if amount <= 0:
                    print("Please enter an amount greater than zero.")
                    continue

                return amount

            except ValueError:
                print("Invalid input. Please enter a valid number.")


# =============================================================
# Creating Customer Objects
# =============================================================

c1 = Bank("Nirmal",5742952759,5600,"Akash","DFC34414")
c2 = Bank("Akash",5242952759,2300,"Akash","DWC34414")
c3 = Bank("Sekher",5762452759,5000,"Akash","DFC34414")

````
