## **Nested `if` Statement**

* When one `if` statement is written **inside another `if` statement**, it is called a **Nested `if`**.
* It is used when one condition is **dependent on another condition**.
* A nested `if` can contain `if`, `else`, or `elif` blocks.
* Proper **indentation** is required to define the nested blocks.

**Syntax:**

```python
if condition:
    if condition:
        statement
    else:
        statement
elif condition:
    if condition:
        statement
else:
    statement
```

````python
age =eval(input("Enter Age:"))
num = int(input("Enter 1 or 0"))

if age >= 18:
    if num == 1:
        print("Eligible to vote")
    else:
        print("Not a citizen")
else:
    print("Underage")
````
