## **Functions**

* A **function** is a reusable block of code designed to perform a **specific task**.
* Functions help divide a program into **small, manageable units**.
* A function can accept **inputs (parameters)** and may return an **output (value)**.
* Functions are defined using the `def` keyword and executed by **calling the function**.

### **Applications**

1. **Reusability:** Write code once and use it multiple times.
2. **Readability:** Makes the program easier to understand.
3. **Structure:** Provides a clear structure to the program.
4. **Maintainability:** Makes code easier to modify, debug, and maintain.
5. **Modularity:** Divides a large program into smaller, independent modules.

## **Types of Functions**

There are two main types of functions in Python:

### **1. Built-in Functions**

* **Built-in functions** are predefined functions provided by Python.
* They are already implemented and can be directly used in a program.
* We can **call and use** them, but we generally do not modify their built-in implementation.

### **Types of Built-in Functions**

1. **Utility Functions:**
   `id()`, `len()`, `type()`, `print()`, `input()`, etc.

2. **String Functions/Methods:**
   Functions or methods used to perform operations on strings.

3. **List Functions/Methods:**
   Functions or methods used to perform operations on lists.

4. **Tuple Functions/Methods:**
   Functions or methods used to perform operations on tuples.

5. **Set Functions/Methods:**
   Functions or methods used to perform operations on sets.

6. **Dictionary Functions/Methods:**
   Functions or methods used to perform operations on dictionaries.

> **Note:** In Python, operations such as `str.upper()`, `list.append()`, and `dict.keys()` are technically called **methods**, because they belong to specific objects/types. Functions like `len()` and `type()` are built-in functions.


### 1. Check whether the given character is a special character or not

```python
ch = input("Enter a character: ")

if ch.isalnum():
    print("Not a special character")
else:
    print("Special character")
```

### 2. Extract uppercase, lowercase, numbers and special characters into different lists

```python
s = 'ADmin@123'

upper = []
lower = []
numbers = []
special = []

for ch in s:
    if ch.isupper():
        upper.append(ch)
    elif ch.islower():
        lower.append(ch)
    elif ch.isdigit():
        numbers.append(ch)
    else:
        special.append(ch)

print("Uppercase:", upper)
print("Lowercase:", lower)
print("Numbers:", numbers)
print("Special characters:", special)
```

### 3. Extract palindrome words from a list and keep them in uppercase in another list

```python
l1 = ['hello', 'level', 'madam', 'python']

out = []

for word in l1:
    if word == word[::-1]:
        out.append(word.upper())

print(out)
```
