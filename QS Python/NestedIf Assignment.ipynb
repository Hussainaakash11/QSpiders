## 1. Divisibility by 2 and 4

```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    if num % 4 == 0:
        print("Divisible by 4")
    else:
        print("Even but not divisible by 4")
else:
    print("Odd number")
```

## 2. ATM Cash Withdrawal

```python
balance = float(input("Enter account balance: "))
amount = float(input("Enter amount to withdraw: "))

if amount <= balance:
    if amount % 100 == 0:
        print("Transaction Successful")
    else:
        print("Enter amount in multiples of 100")
else:
    print("Insufficient Balance")
```

## 3. Palindrome only if the data is a string

```python
data = input("Enter data: ")

if data.isalpha():
    if data == data[::-1]:
        print("Palindrome")
    else:
        print("Not a palindrome")
else:
    print("The given data is not a string.")
```

## 4. Reverse data only if it has more than 3 values

```python
data = input("Enter data: ")

if len(data) > 3:
    print("Reversed data:", data[::-1])
else:
    print("Data does not have more than 3 values.")
```

## 5. Vowel or consonant

```python
char = input("Enter a character: ")

if len(char) == 1 and char.isalpha():
    if char.lower() in "aeiou":
        print("Vowel")
    else:
        print("Consonant")
else:
    print("Enter a single alphabet.")
```

## 6. Smallest among three numbers using nested if

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))

if a <= b:
    if a <= c:
        smallest = a
    else:
        smallest = c
else:
    if b <= c:
        smallest = b
    else:
        smallest = c

print("Smallest number =", smallest)
```

## 7. Largest among three numbers using nested if

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))

if a >= b:
    if a >= c:
        largest = a
    else:
        largest = c
else:
    if b >= c:
        largest = b
    else:
        largest = c

print("Largest number =", largest)
```

## 9. Loan Eligibility Check

```python
age = int(input("Enter age: "))
income = float(input("Enter monthly income: ₹"))
credit_score = int(input("Enter credit score: "))

if age >= 21:
    if income >= 30000:
        if credit_score >= 700:
            print("Customer is eligible for the loan.")
        else:
            print("Not eligible: Credit score must be at least 700.")
    else:
        print("Not eligible: Monthly income must be at least ₹30,000.")
else:
    print("Not eligible: Age must be at least 21.")
```

## 10. E-commerce Discount System

```python
purchase = float(input("Enter purchase amount: ₹"))
premium = input("Are you a premium member? (yes/no): ").lower()

if purchase > 10000:
    discount = 20
elif purchase > 5000:
    discount = 10
else:
    discount = 0

if premium == "yes":
    discount = discount + 5

discount_amount = purchase * discount / 100
final_bill = purchase - discount_amount

print("Discount =", discount, "%")
print("Discount amount = ₹", discount_amount)
print("Final bill amount = ₹", final_bill)
```

## 11. Cab Fare Surge Pricing

```python
base_fare = 100

peak = input("Is it peak hour? (yes/no): ").lower()
rainy = input("Is the weather rainy? (yes/no): ").lower()
holiday = input("Is it a holiday? (yes/no): ").lower()

fare = base_fare

if peak == "yes":
    fare = fare + (base_fare * 50 / 100)

if rainy == "yes":
    fare = fare + (base_fare * 20 / 100)

if holiday == "yes":
    fare = fare + (base_fare * 30 / 100)

print("Final cab fare = ₹", fare)
```
