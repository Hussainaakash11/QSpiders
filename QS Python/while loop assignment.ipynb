## 13. Check Armstrong number

```python
n = int(input("Enter number: "))
original = n
power = len(str(n))
total = 0

while n > 0:
    digit = n % 10
    total = total + digit ** power
    n = n // 10

if total == original:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

## 14. Check Spy number

```python
n = int(input("Enter number: "))
total = 0
product = 1

while n > 0:
    digit = n % 10
    total = total + digit
    product = product * digit
    n = n // 10

if total == product:
    print("Spy number")
else:
    print("Not a Spy number")
```

## 15. Check Xylem or Phloem number

```python
n = int(input("Enter number: "))

last = n % 10
n = n // 10
middle = 0

while n > 9:
    digit = n % 10
    middle = middle + digit
    n = n // 10

first = n

if first + last == middle:
    print("Xylem number")
else:
    print("Phloem number")
```

## 16. Check Disarium number

```python
n = int(input("Enter number: "))
original = n
power = len(str(n))
position = power
total = 0

while n > 0:
    digit = n % 10
    total = total + digit ** position
    position = position - 1
    n = n // 10

if total == original:
    print("Disarium number")
else:
    print("Not a Disarium number")
```

## 20. Check Strong number

```python
n = int(input("Enter number: "))
original = n
total = 0

while n > 0:
    digit = n % 10

    fact = 1
    i = 1

    while i <= digit:
        fact = fact * i
        i = i + 1

    total = total + fact
    n = n // 10

if total == original:
    print("Strong number")
else:
    print("Not a Strong number")
```

## 24. Extract palindrome numbers from a list

## 30. Extract prime numbers from a list

```python
numbers = [2, 4, 5, 7, 10, 11, 13, 15, 17]
result = []
i = 0

while i < len(numbers):
    n = numbers[i]
    divisor = 1
    count = 0

    while divisor <= n:
        if n % divisor == 0:
            count = count + 1
        divisor = divisor + 1

    if count == 2:
        result.append(n)

    i = i + 1

print("Prime numbers =", result)
```

## 31. Extract Armstrong numbers from a list

```python
numbers = [153, 10, 370, 25, 407, 123]
result = []
i = 0

while i < len(numbers):
    n = numbers[i]
    original = n
    power = len(str(n))
    total = 0

    while n > 0:
        digit = n % 10
        total = total + digit ** power
        n = n // 10

    if total == original:
        result.append(original)

    i = i + 1

print("Armstrong numbers =", result)
```

## 32. Extract Perfect numbers from a tuple

```python
numbers = (6, 10, 28, 12, 496, 20)
result = []
i = 0

while i < len(numbers):
    n = numbers[i]
    divisor = 1
    total = 0

    while divisor < n:
        if n % divisor == 0:
            total = total + divisor
        divisor = divisor + 1

    if total == n:
        result.append(n)

    i = i + 1

print("Perfect numbers =", result)
```

## 33. Extract Xylem numbers from a list

```python
numbers = [1234, 123, 246, 121, 12345]
result = []
i = 0

while i < len(numbers):
    n = numbers[i]

    last = n % 10
    temp = n // 10
    middle = 0

    while temp > 9:
        middle = middle + temp % 10
        temp = temp // 10

    first = temp

    if first + last == middle:
        result.append(n)

    i = i + 1

print("Xylem numbers =", result)
```
