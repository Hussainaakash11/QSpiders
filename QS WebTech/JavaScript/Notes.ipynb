## **DAY 2**

### **1. var keyword:**
    - Multiple time declaration.
    - Multiple time initilization.
    - variable decleared with `var` Keyword goes to global scope.
    - variable decleared with `var` Keyword supports hoisting and does not belong to temporal dead zone (TDZ).

### **2. let keyword:**
    - One time declaration.
    - Multiple time initilization.
    - variable decleared with `let` Keyword goes to block scope.
    - variable decleared with `let` Keyword supports hoisting and belong to temporal dead zone (TDZ).

### **3. const keyword:**
    - One time declaration.
    - One time initilization.
    - variable decleared with `const` Keyword goes to block scope or script scope.
    - variable decleared with `const` Keyword supports hoisting and belong to temporal dead zone (TDZ).

## **DAY 5**

- When the value of a variable is not present in the local/block scope, the JavaScript engine searches outward through the outer lexical environments (such as the script scope and then the global scope). Once the block finishes execution, unreferenced block-scoped variables become eligible for garbage collection. This hierarchical variable resolution mechanism is known as **lexical scoping** (and the traversal path is the **scope chain**).

### **Functions**
- A function is a reusable block of code designed to perform a specific task.
- The main advantage of functions is achieving code reusability and modularity.
- To call (invoke) a function, we need two things: the function name and parentheses `()`.
- The values passed during a function call statement are known as **arguments**.
- The variables declared in a function definition are known as **parameters**.
- Variables declared inside a function (even with the `var` keyword) belong to the **local (function) scope**.
- Syntax:
```javascript
// function declearation
function function_name(par1,par2)
{
    // function block
}
// function call
function_name(arg1,arg2)
```
> **Note:** If a function does not return anything that means it returns undefined.

### **Default Parameters:**
- If we dont pass any arg to a function it stores undefined by default.
- To overcome this issue we use default parameters.
